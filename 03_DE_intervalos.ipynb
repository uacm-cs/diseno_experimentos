# Distribuciones de Probabilidad para Experimentos


## ¿Qué es una Distribución de Probabilidad?

### Una **distribución de probabilidad** describe los posibles valores de una **variable aleatoria** y las probabilidades asociadas a esos valores.

### Ejemplo: Tiempo de ejecución

### Supongamos que medimos el tiempo de ejecución de un programa en 1000 ocasiones. La distribución nos dice:
### - ¿Qué valores son más frecuentes?
### - ¿Qué tan dispersos están?
### - ¿Qué tan probable es observar un valor específico?

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
````

````python
# ============================================================
# SIMULACIÓN DE TIEMPOS DE EJECUCIÓN
# ============================================================
# En este ejemplo, simulamos 1000 mediciones del tiempo de
# ejecución de un programa. Usamos una distribución normal
# con media 100 ms y desviación estándar 10 ms.
#
# ---- Paso 1: Fijar la semilla de aleatoriedad ----
# np.random.seed() fija la semilla para que los resultados
# sean REPRODUCIBLES. Si se usa la misma semilla (42), obtendrás
# los mismos números aleatorios cada vez que ejecutes el código.
#   - Permite que otros investigadores reproduzcan tus resultados.
#   - Es una buena práctica en experimentos computacionales.

np.random.seed(42)

# ---- Paso 2: Simular los tiempos de ejecución ----
# np.random.normal() genera números aleatorios que siguen
# una distribución normal.
#
# Parámetros:
#   - loc=100: la media (μ) de la distribución. En este caso,
#     el tiempo promedio de ejecución es 100 ms.
#   - scale=10: la desviación estándar (σ). En este caso,
#     los tiempos varían típicamente 10 ms alrededor de la media.
#   - size=1000: el número de observaciones a generar.
#     Simulamos 1000 ejecuciones del programa.
#
# Resultado:
#   - 'tiempos' es un arreglo (array) de NumPy con 1000 valores.

tiempos = np.random.normal(loc=100, scale=10, size=1000000)

# ============================================================
# CALCULAR ESTADÍSTICOS DESCRIPTIVOS
# ============================================================

# ---- Media ----
print(f"Media: {np.mean(tiempos):.2f} ms")

# ---- Desviación estándar ----
print(f"Desviación estándar: {np.std(tiempos, ddof=1):.2f} ms")

# ---- Mínimo ----
print(f"Mínimo: {np.min(tiempos):.2f} ms")

# ---- Máximo ----
print(f"Máximo: {np.max(tiempos):.2f} ms")
````

````python
# Visualizar la distribución empírica
plt.figure(figsize=(10, 6))
plt.hist(
    tiempos,              # 1. Los datos a graficar (lista o array)
    bins=100,             # 2. Número de intervalos (barras)
    edgecolor='blue',     # 3. Color del borde de cada barra
    color='steelblue',    # 4. Color de relleno de las barras
    alpha=0.7,            # 5. Transparencia de las barras (0 a 1)
    density=True          # 6. Normaliza el área total a 1
)
plt.title('Distribución empírica de tiempos de ejecución')
plt.xlabel('Tiempo (ms)')
plt.ylabel('Densidad')
plt.grid(True, alpha=0.3)
plt.show()
````

##  Distribución Empírica vs. Distribución Teórica

| **Tipo** | **¿Qué es?** | **¿Cómo se obtiene?** |
| :--- | :--- | :--- |
| **Empírica** | Basada en datos observados | Histograma de los datos |
| **Teórica** | Modelo matemático | Fórmula (ej: distribución normal) |



````python
# ============================================================
# COMPARACIÓN: DISTRIBUCIÓN EMPÍRICA VS. TEÓRICA
# ============================================================
# OBJETIVO:
#   Superponer el histograma de los datos simulados (distribución
#   empírica) con la curva de la distribución normal teórica
#   para comparar ambas.
#
# ¿POR QUÉ ES ÚTIL?
#   - Nos permite ver si los datos se aproximan a la normal.
#   - Nos ayuda a evaluar si el modelo teórico es razonable.
#   - Es una forma visual de verificar supuestos.
# ============================================================

# ---- Paso 1: Crear la figura ----
plt.figure(figsize=(10, 6))

# ---- Paso 2: Dibujar el histograma (distribución empírica) ----
# plt.hist() dibuja el histograma de los datos observados.
# Parámetros:
#   - tiempos: los datos simulados (1000 valores).
#   - bins=30: divide el rango de datos en 30 intervalos.
#   - density=True: normaliza el histograma para que el área
#     total sea 1. Esto permite compararlo con una función
#     de densidad teórica (como la normal).
#   - label='Empírica': etiqueta para la leyenda.
plt.hist(tiempos, bins=30, edgecolor='black', color='steelblue',
         alpha=0.5, density=True, label='Empírica')

# ---- Paso 3: Crear la curva normal teórica ----
# np.linspace() genera un conjunto de valores equidistantes.
#   - 60: valor inicial del rango.
#   - 140: valor final del rango.
#   - 1000: número de puntos a generar.
# Esto crea 1000 valores entre 60 y 140 para dibujar la curva.
x = np.linspace(60, 140, 1000)

# stats.norm.pdf() calcula la FUNCIÓN DE DENSIDAD DE PROBABILIDAD
# (PDF, Probability Density Function) de una distribución normal.
# Parámetros:
#   - x: los valores donde evaluar la PDF.
#   - loc=100: media de la distribución normal (μ).
#   - scale=10: desviación estándar (σ).
# Resultado:
#   - y: los valores de densidad correspondientes a cada x.
y = stats.norm.pdf(x, loc=100, scale=10)

# ---- Paso 4: Dibujar la curva normal teórica ----
# plt.plot() dibuja una línea continua.
# Parámetros:
#   - x: valores del eje X.
#   - y: valores del eje Y (densidad).
#   - 'r-': formato de línea: 'r' = rojo, '-' = línea continua.
#   - linewidth=2: grosor de la línea.
#   - label='Normal teórica (μ=100, σ=10)': etiqueta para la leyenda.
plt.plot(x, y, 'r-', linewidth=2, label='Normal teórica (μ=100, σ=10)')

# ---- Paso 5: Personalizar el gráfico ----
plt.title('Distribución empírica vs. teórica')
plt.xlabel('Tiempo (ms)')
plt.ylabel('Densidad')
plt.legend()

# plt.grid() activa la cuadrícula.
#   - True: muestra la cuadrícula.
#   - alpha=0.3: transparencia (0=invisible, 1=opaco).
plt.grid(True, alpha=0.3)

plt.show()
````

## Ejemplo: Tiempo de compilación en un pipeline CI/CD

### Contexto

#### El equipo de DevOps de una empresa ha medido históricamente el tiempo que tarda en compilar y empaquetar su microservicio principal. Han determinado que este tiempo sigue una **distribución normal** con una media ($\mu$) de **50 segundos** y una desviación estándar ($\sigma$) de **5 segundos**.

#### El equipo de QA quiere analizar la probabilidad de que el proceso de compilación tarde **entre 42 y 52 segundos**, ya que ese es el rango de tiempo que tienen asignado en su pipeline de Integración Continua (CI) antes de que se active una alerta de cuello de botella.

![Distribución normal del tiempo de compilación](./figs/fig-normal_ejem1.png)

### ¿Cómo se conecta con el estadístico \(Z\)?

### Se **estandarizamos** los valores.

### El estadístico \(Z\) mide **cuántas desviaciones estándar** se aleja un valor de la media:

$$
Z = \frac{X - \mu}{\sigma} 
$$

### **Paso 1: Estandarizar el límite inferior (42 s)**
$$
 Z_1 = \frac{42 - 50}{5} = \frac{-8}{5} = -1.60 
$$

### **Paso 2: Estandarizar el límite superior (52 s)**
$$
 Z_2 = \frac{52 - 50}{5} = \frac{2}{5} = 0.40 
$$

### **Paso 3: Conectar con el área (probabilidad)**
### Ahora que tenemos los valores en la escala $Z$, buscamos el área bajo la curva normal estándar entre $Z = -1.60$ y $Z = 0.40$. 

### En Python, esto se traduce en:
### 1. Calcular el área acumulada hasta $Z_2 = 0.40$.
### 2. Calcular el área acumulada hasta $Z_1 = -1.60$.
### 3. Restar ambas áreas.

$$
 P(42 \leq X \leq 52) = P(-1.60 \leq Z \leq 0.40) = \Phi(0.40) - \Phi(-1.60) 
$$

> **💡 Nota:** $\Phi$ es la función de distribución acumulada de la normal estándar (en Python: `stats.norm.cdf`).

````python
# ============================================================
# CÁLCULO DE PROBABILIDAD CON LA DISTRIBUCIÓN NORMAL
# ============================================================
# ---- Paso 1: Importar la librería ----
# scipy.stats contiene las funciones estadísticas y las
# distribuciones de probabilidad (normal, t, chi-cuadrada, etc.).
import scipy.stats as stats

# ---- Paso 2: Definir los parámetros de la distribución ----
mu = 50           # mu: Media en segundos
sigma = 5         # sigma: Desviación estándar en segundos

# ---- Paso 3: Definir los límites del intervalo ----
# x_inferior: límite inferior del rango de interés (42 s).
x_inferior = 42   # Límite inferior

# x_superior: límite superior del rango de interés (52 s).
x_superior = 52   # Límite superior

# Cálculo de la puntuación z (z-score)
z_inferior = (x_inferior - mu) / sigma
z_superior = (x_superior - mu) / sigma
print(f"Z1 (para 42s): {z_inferior:.2f}")
print(f"Z2 (para 52s): {z_superior:.2f}")

# ============================================================
# PASO 4: CALCULAR EL ÁREA BAJO LA CURVA (PROBABILIDAD)
# ============================================================
# La función stats.norm.cdf(x, loc, scale) calcula la
# PROBABILIDAD ACUMULADA desde -infinito hasta x.
# Esto es, devuelve la probabilidad de que una variable aleatoria normal 
# sea menor o igual a un valor dado x P(X ≤ x).
#   - x: el valor hasta donde queremos acumular.
#   - loc: la media (mu) de la distribución, default 0.
#   - scale: la desviación estándar (sigma), default 1.

# Para los valores z se aplica con la distribución normal estándar N(mu=0,sigma=1)
area_z_superior = stats.norm.cdf(z_superior)

# Área acumulada hasta el límite inferior (42 s).
# Representa la probabilidad de que el tiempo sea <= 42 s.
area_z_inferior = stats.norm.cdf(z_inferior)

probabilidad = area_z_superior - area_z_inferior

print(f"Probabilidad de que la compilación tarde entre {z_inferior}s y {z_superior}s: {probabilidad*100:.2f}%")
````

````python
# ============================================================
# OTRA FORMA DE CALCULAR EL ÁREA BAJO LA CURVA (PROBABILIDAD)
# ============================================================
# La función stats.norm.cdf(x, loc, scale) calcula la
# PROBABILIDAD ACUMULADA desde -infinito hasta x.
# Esto es, devuelve la probabilidad de que una variable aleatoria normal 
# sea menor o igual a un valor dado x P(X ≤ x).
#   - x: el valor hasta donde queremos acumular.
#   - loc: la media (mu) de la distribución, default 0.
#   - scale: la desviación estándar (sigma), default 1.

# Representa la probabilidad de que el tiempo sea <= 52 s.
area_hasta_52 = stats.norm.cdf(x_superior, loc=mu, scale=sigma)

# Área acumulada hasta el límite inferior (42 s).
# Representa la probabilidad de que el tiempo sea <= 42 s.
area_hasta_42 = stats.norm.cdf(x_inferior, loc=mu, scale=sigma)

# ---- Paso 5: Calcular el área entre los dos límites ----
# La probabilidad de estar ENTRE 42 y 52 es la diferencia:
#   Área(<= 52) - Área(<= 42)
probabilidad = area_hasta_52 - area_hasta_42

print(f"Probabilidad de que la compilación tarde entre {x_inferior}s y {x_superior}s: {probabilidad*100:.2f}%")

````

# Nivel de Confianza y $\alpha$

- ### El cálculo del área entre 42 y 52 segundos nos dio una probabilidad del 60.06%. 
- ### Pero, ¿cómo se relaciona esto con los **niveles de confianza** y el **estadístico \(Z\)**?

## 1. El Nivel de Confianza ($1 - \alpha$)
### El nivel de confianza es el **área central** bajo la curva. Si queremos un 95% de confianza, el área central es 0.95. El 5% restante se reparte en las dos colas (2.5% en cada una).

## 2. El Nivel de Significancia ($\alpha$)
### $\alpha$ es el área de las **colas** (los extremos). Para un 95% de confianza, $\alpha = 0.05$. Es el riesgo que aceptamos de que el valor caiga fuera del rango.

## 3. El Estadístico \(Z\) (Estandarización)
- ### El **estadístico \(Z\)** es un número que nos indica **a cuántas desviaciones estándar** se encuentra un valor específico respecto a la media. 
- ### Su función es **traducir** nuestros datos originales (en segundos) a una **escala común y estandarizada** (la curva normal estándar, con media 0 y desviación 1).

### La fórmula para calcularlo es:
$$
 Z = \frac{X - \mu}{\sigma} 
$$

### Donde:
- ####    **$X$**: Es el valor original que queremos estandarizar (por ejemplo, los 42 o 52 segundos).
- ####    **$\mu$**: Es la media de la distribución (en nuestro ejemplo, 50 s).
- ####    **$\sigma$**: Es la desviación estándar de la distribución (en nuestro ejemplo, 5 s).


## 4. El Estadístico $Z$ Crítico
- ### El **$Z$ crítico** es un valor fijo que se busca en la tabla de la normal estándar (o se calcula con software). 
- ### Es el valor en el eje X que separa la zona central (aceptación) de las colas (rechazo). 
- ### Para un 95% de confianza, el $Z$ crítico es **$\pm 1.96$**.

> ###  **Nota:** No confundir el **estadístico $Z$** (el valor que se calcula con los datos, como -1.60 o 0.40) con el **$Z$ crítico** (el valor fijo de la tabla para un nivel de confianza, como $ \pm 1.96 $).

## 5. Convertir $Z$ Crítico a Segundos (Desestandarizar)
### Para encontrar los límites en segundos (el Intervalo de Confianza), usamos el $Z$ crítico y despejamos la fórmula:
$$
 \text{Límite} = \mu \pm (Z_{\text{crítico}} \times \sigma) 
$$
$$
 \text{Límite} = 50 \pm (1.96 \times 5) 
$$

- ###   **Límite inferior:** $50 - 9.8 = 40.2$ s
- ###   **Límite superior:** $50 + 9.8 = 59.8$ s

> ### **Interpretación:** El rango de 42 a 52 s que analizamos antes solo cubría el 60% de los casos. Si usamos un nivel de confianza del 95%, el rango se amplía a **40.2 - 59.8 s**.

````python
# ============================================================
# ÁREA BAJO LA CURVA -> Z CRÍTICO -> INTERVALO DE CONFIANZA
# ============================================================
# En este código calculamos:
# 1. El Intervalo de Confianza (IC) al 95% para el tiempo de compilación.
# 2. Comparamos el área de ese IC con el área del rango original (42-52 s).
# ============================================================

# ---- Paso 1: Importar la librería estadística ----
# scipy.stats contiene las funciones de distribución de probabilidad.
import scipy.stats as stats

# ---- Paso 2: Definir los parámetros de la población ----
# Son los datos conocidos del contexto de DevOps.
mu = 50       # Media poblacional del tiempo de compilación (segundos)
sigma = 5     # Desviación estándar poblacional (segundos)

# ============================================================
# BLOQUE 1: CALCULAR LOS LÍMITES PARA UN NIVEL DE CONFIANZA DEL 95%
# ============================================================

# ---- Paso 3: Definir el nivel de confianza y alpha ----
confianza = 0.95        # Queremos un 95% de confianza
alpha = 1 - confianza   # El nivel de significancia es 0.05 (el 5% restante)

# ---- Paso 4: Calcular el valor Z crítico ----
# Usamos stats.norm.ppf() que nos da el valor Z acumulado a la izquierda.
# ============================================================
# stats.norm.ppf(q, loc=0, scale=1)
# ============================================================
# q     : área acumulada a la izquierda (obligatorio)
# loc   : media (por defecto 0)
# scale : desviación estándar (por defecto 1)
#
# Con los valores por defecto, la función trabaja en la
# escala Z (normal estándar).
# ============================================================

# Para un intervalo central, buscamos el área (1 - alpha/2).
# alpha/2 = 0.025, entonces 1 - 0.025 = 0.975.
# ppf(0.975) devuelve 1.96 (el famoso valor crítico para 95% de confianza).

z_critico = stats.norm.ppf(1 - alpha/2)
print(f"Z crítico para {confianza*100}% de confianza: ±{z_critico:.2f}")

# ---- Paso 5: Desestandarizar (convertir Z a segundos) ----
# El Z crítico está en la escala estandarizada (0 y 1).
# Para volver a los segundos, usamos la fórmula inversa:
# Límite = Media ± (Z crítico * Desviación Estándar)
limite_inferior_95 = mu - (z_critico * sigma)
limite_superior_95 = mu + (z_critico * sigma)

print(f"Intervalo de Confianza al 95%: [{limite_inferior_95:.2f}s, {limite_superior_95:.2f}s]")

# ============================================================
# BLOQUE 2: VERIFICAR EL ÁREA DEL INTERVALO ORIGINAL (42 A 52)
# ============================================================

# ---- Paso 6: Calcular los valores Z para los límites originales ----
# Convertimos los límites en segundos (42 y 52) a la escala Z.
# Z = (X - mu) / sigma
z_42 = (42 - mu) / sigma   # Z para el límite inferior (42 s)
z_52 = (52 - mu) / sigma   # Z para el límite superior (52 s)

# ---- Paso 7: Calcular el área entre los dos límites ----
# stats.norm.cdf() calcula el área acumulada a la izquierda del valor Z.
# El área entre dos valores es la diferencia de sus áreas acumuladas.
area_hasta_52 = stats.norm.cdf(z_52)
area_hasta_42 = stats.norm.cdf(z_42)

area_42_52 = area_hasta_52 - area_hasta_42

# ---- Paso 8: Mostrar el resultado en porcentaje ----
# Multiplicamos por 100 para obtener el porcentaje y redondeamos a 2 decimales.
print(f"\nEl rango original [42s, 52s] cubre el {area_42_52*100:.2f}% de los casos.")
````

## Ejemplo: Tiempo medio de resolución de bugs

### Contexto
### Un equipo de QA quiere estimar el **tiempo medio real** que tardan los desarrolladores en resolver un bug usando una nueva herramienta de depuración. 

### Como no pueden medir todos los bugs del mundo (población), tomaron una muestra de **25 bugs** ($n = 25$). 
- ### La media muestral fue de **40 minutos** ($\bar{x} = 40$).
- ### La desviación estándar muestral fue de **10 minutos** (\(s = 10\)).

### Objetivo
### Calcular un **Intervalo de Confianza del 95%** para el tiempo medio de resolución. 

$$
\Large
 IC_{95\%} = \bar{x} \pm \left( t_{1 - \alpha/2, \, df} \times \frac{s}{\sqrt{n}} \right) 
$$

### Representa el error estándar (SE) : $\frac{s}{\sqrt{n}}$



> ### **¿Por qué usamos la distribución $t$ y no la normal?**
> ### Porque no conocemos la desviación estándar poblacional ($\sigma$). Solo conocemos la muestral ($s$). La distribución $t$ de Student es la adecuada para esta situación.

````python
# ============================================================
# CÁLCULO DE UN INTERVALO DE CONFIANZA CON LA DISTRIBUCIÓN t
# ============================================================
# OBJETIVO:
#   Calcular el intervalo de confianza del 95% para la media
#   de un experimento, usando la distribución t de Student
#   (porque no conocemos la desviación estándar poblacional).
# ============================================================

# ---- Paso 0: Importar las librerías necesarias ----
# scipy.stats contiene las funciones de distribución de
# probabilidad (como la t de Student).
import scipy.stats as stats

# numpy contiene funciones matemáticas (como la raíz cuadrada).
import numpy as np


# ============================================================
# DATOS DEL PROBLEMA
# ============================================================
# Estos son los datos que conocemos de nuestro experimento.
# En un caso real, vendrían de haber medido la muestra.

n = 25             # Tamaño de muestra (cuántos datos tenemos)
x_bar = 40         # Media muestral (el promedio de los datos)
s = 10              # Desviación estándar muestral (la dispersión)
confianza = 0.95    # Nivel de confianza que queremos (95%)

# El nivel de significancia (alpha) es el complemento de la confianza.
# Si queremos 95% de confianza, alpha = 0.05 (el 5% de riesgo).
alpha = 1 - confianza


# ============================================================
# PASO 1: Calcular los grados de libertad
# ============================================================
# Para una muestra, los grados de libertad son n - 1.
# Recuerda: la media "consume" un grado de libertad.
df = n - 1
print(f"Grados de libertad (df): {df}")


# ============================================================
# PASO 2: Buscar el valor crítico t
# ============================================================
# stats.t.ppf() nos da el valor t acumulado a la izquierda.
# Como queremos un intervalo central, dejamos alpha/2 en cada cola.
# Para 95%: alpha/2 = 0.025, entonces buscamos ppf(0.975).
t_critico = stats.t.ppf(1 - alpha/2, df)
print(f"Valor crítico t: {t_critico:.4f}")


# ============================================================
# PASO 3: Calcular el error estándar
# ============================================================
# El error estándar mide la variabilidad de la media muestral.
# Fórmula: s / sqrt(n)
error_estandar = s / np.sqrt(n)
print(f"Error estándar: {error_estandar:.4f}")


# ============================================================
# PASO 4: Calcular el margen de error
# ============================================================
# El margen de error es cuánto sumamos y restamos a la media.
# Fórmula: t_critico * error_estandar
margen_error = t_critico * error_estandar
print(f"Margen de error: {margen_error:.4f}")


# ============================================================
# PASO 5: Construir el intervalo
# ============================================================
# El intervalo va desde (media - margen) hasta (media + margen).
limite_inferior = x_bar - margen_error
limite_superior = x_bar + margen_error

print(f"\nIC 95%: [{limite_inferior:.2f}, {limite_superior:.2f}]")
````

## Ejercicio 3: Latencia de una API

### Contexto
### Un equipo de backend está monitoreando la latencia de una API crítica que da servicio a una aplicación web. Para evaluar su rendimiento, el equipo tomó una muestra aleatoria de 15 mediciones de latencia (en milisegundos) durante una hora de operación normal.
### El equipo solo tiene 15 mediciones (n=15) y no conoce la desviación estándar poblacional de la latencia para estimar el intervalo de confianza del tiempo medio real de respuesta.


### Datos

### Muestra de 15 mediciones (ms):

```python
latencias = [32.1, 28.6, 22.1, 50.9, 32.4, 28.2, 22.5, 50.2, 32.1, 28.4, 22.6, 50.7, 32.7, 28.7, 22.2]
```
