## ¿Qué son las pruebas de hipótesis?

### Son procedimientos estadísticos que nos permiten **tomar decisiones** sobre una población a partir de los datos de una muestra. Nos ayudan a responder preguntas como:

> ### *"¿La nueva herramienta reduce realmente el tiempo de desarrollo, o la diferencia observada es solo producto del azar?"*

## Objetivos

### 1. Entender qué es una prueba de hipótesis.
### 2. Diferenciar las pruebas paramétricas de las no paramétricas.
### 3. Identificar cuándo usar cada una según los supuestos de los datos.
### 4. Conocer las pruebas más comunes de cada familia.

# 1. La decisión fundamental: usar una prueba ¿paramétrica o no paramétrica?

### Antes de aplicar cualquier prueba, debemos hacernos una pregunta clave: **¿mis datos cumplen los supuestos?**

| <span style="font-size: 28px;">**Tipo de prueba**</span> | <span style="font-size: 28px;">**¿Qué asume?**</span> | <span style="font-size: 28px;">**¿Qué usa?**</span> | <span style="font-size: 28px;">**¿Cuándo usarla?**</span> |
| :--- | :--- | :--- | :--- |
| <span style="font-size: 28px;"> **Paramétrica**</span> | <span style="font-size: 28px;"> Normalidad y (usualmente) homocedasticidad.</span>  | <span style="font-size: 28px;">Medias y varianzas. </span>| <span style="font-size: 28px;">Cuando los datos cumplen los supuestos.</span> |
| <span style="font-size: 28px;">**No paramétrica**</span> | <span style="font-size: 28px;">No asume una distribución específica. | <span style="font-size: 28px;">Rangos u ordenaciones.</span> | <span style="font-size: 28px;">Cuando los datos no son normales, son ordinales o hay valores atípicos.</span> |

</br>

> ### **Importante:** Si los datos cumplen los supuestos, usar la prueba paramétrica (es más potente). Si no, usa la no paramétrica (es más segura).

## 2. El Mapa de las Pruebas de Hipótesis

### La siguiente tabla resume las pruebas más comunes según el **objetivo**, el **número de grupos** y el **tipo de diseño** (independiente o pareado):

| **Objetivo del análisis** | **Número de grupos** | **Prueba Paramétrica** | **Prueba No Paramétrica** |
| :--- | :---: | :--- | :--- |

| <p style="color:DodgerBlue;"> **Comparar medias**  </p> | <p style="color:DodgerBlue;"> 3+ (pareados) </p>| <p style="color:DodgerBlue;"> ANOVA de medidas repetidas </p>| <p style="color:DodgerBlue;"> **Friedman** </p>|

# ANOVA de Medidas Repetidas con `pingouin`

### `pingouin.rm_anova(data=None, dv=None, within=None, subject=None, correction='auto', detailed=False, effsize='ng2')`

* #### Contrasta si las **medias poblacionales** de **tres o más condiciones pareadas** (medidas repetidas sobre la misma unidad experimental) son iguales.
* #### Es la generalización del ANOVA de un factor al caso en que las observaciones **no son independientes**: cada sujeto se mide en todas las condiciones.
* #### `pingouin` calcula automáticamente la **prueba de esfericidad de Mauchly** y, si se viola, aplica la corrección de **Greenhouse-Geisser** al valor p.

## ¿Qué significa "pareado" o "medidas repetidas" aquí?

* #### **Medidas repetidas** significa que cada unidad experimental (el mismo desarrollador, el mismo módulo, el mismo equipo) se mide en **todas** las condiciones del factor.
* #### No son grupos de sujetos distintos, sino **el mismo sujeto expuesto a todas las condiciones**.
* #### Se usa cuando las mediciones provienen de **las mismas unidades** en distintas condiciones (tres lenguajes implementados por los mismos desarrolladores, tres arquitecturas de Transformer evaluadas sobre los mismos datasets, tres metodologías probadas por los mismos equipos, etc.).

## ¿En qué consiste el ANOVA de medidas repetidas?

* #### El ANOVA de medidas repetidas **no compara las condiciones par a par**, sino que contrasta **globalmente** si todas las medias son iguales.
* #### Descompone la **varianza total** en tres fuentes:
  * #### **Varianza entre condiciones** ($MS_{\text{condiciones}}$): refleja cuánto difieren las medias de las condiciones.
  * #### **Varianza entre sujetos** ($MS_{\text{sujetos}}$): refleja las diferencias individuales entre sujetos (se controla, no es de interés).
  * #### **Varianza de error** ($MS_{\text{error}}$): el ruido residual, una vez eliminada la variabilidad entre sujetos.

## El estadístico $F$ es el cociente entre las dos fuentes relevantes:

$$
\large
F = \frac{MS_{\text{condiciones}}}{MS_{\text{error}}}
$$

### donde
* #### $MS_{\text{condiciones}} = \dfrac{SS_{\text{condiciones}}}{k-1}$, con $k$ = número de condiciones.
* #### $MS_{\text{error}} = \dfrac{SS_{\text{error}}}{(k-1)(n-1)}$, con $n$ = número de sujetos.
* #### Bajo $H_0$, este estadístico sigue una distribución $F$ de Fisher con $k-1$ y $(k-1)(n-1)$ grados de libertad.

## 1. Fórmulas

### Sea:
- #### $k$ = número de condiciones (niveles del factor de medidas repetidas).
- #### $n$ = número de sujetos (unidades experimentales).
- #### $x_{ij}$ = observación del sujeto $i$ en la condición $j$.
- #### $\bar{x}_{.j} = \frac{1}{n}\sum_{i=1}^{n} x_{ij}$ = media de la condición $j$.
- #### $\bar{x}_{i.} = \frac{1}{k}\sum_{j=1}^{k} x_{ij}$ = media del sujeto $i$.
- #### $\bar{x} = \frac{1}{nk}\sum_{i=1}^{n}\sum_{j=1}^{k} x_{ij}$ = media global.

### 1.1 Suma de cuadrados de las condiciones

$$
\large
SS_{\text{condiciones}} = n \sum_{j=1}^{k} (\bar{x}_{.j} - \bar{x})^2
$$

#### Mide cuánto se aleja **la media de cada condición** de la media global, ponderando por el número de sujetos.

### 1.2 Suma de cuadrados de los sujetos

$$
\large
SS_{\text{sujetos}} = k \sum_{i=1}^{n} (\bar{x}_{i.} - \bar{x})^2
$$

#### Mide cuánto se aleja **la media de cada sujeto** de la media global. Esta variabilidad **se elimina** del error, lo que hace más potente la prueba.

### 1.3 Suma de cuadrados del error

$$
\large
SS_{\text{error}} = \sum_{i=1}^{n}\sum_{j=1}^{k} (x_{ij} - \bar{x}_{i.} - \bar{x}_{.j} + \bar{x})^2
$$

#### Mide la variabilidad **residual**, una vez eliminados los efectos de condición y de sujeto.

## 2. Grados de libertad

| Término | Grados de libertad | Razón |
| :--- | :---: | :--- |
| $SS_{\text{condiciones}}$ | $k - 1$ | $k$ condiciones menos 1 |
| $SS_{\text{sujetos}}$ | $n - 1$ | $n$ sujetos menos 1 |
| $SS_{\text{error}}$ | $(k-1)(n-1)$ | Residual |
| $SS_{\text{total}}$ | $nk - 1$ | $nk$ observaciones menos 1 |

#### Verificación: $(k-1) + (n-1) + (k-1)(n-1) = nk - 1$

## 3. Cuadrados medios (MS)

$$
\large
MS_{\text{condiciones}} = \frac{SS_{\text{condiciones}}}{k - 1}, \qquad MS_{\text{error}} = \frac{SS_{\text{error}}}{(k-1)(n-1)}
$$

> #### $MS_{\text{condiciones}}$ estima la varianza entre condiciones; $MS_{\text{error}}$ estima el ruido residual.

### Interpretación

- #### Bajo $H_0$ ($\mu_1 = \dots = \mu_k$), **ambos $MS$ estiman la misma varianza** → $F \approx 1$.
- #### Si las medias de las condiciones difieren mucho, $MS_{\text{condiciones}}$ crece → $F$ grande → se rechaza $H_0$.
- #### Bajo $H_0$, $F$ sigue una distribución $F$ de Fisher con $(k-1, (k-1)(n-1))$ grados de libertad.

## Las hipótesis son:

$$
\large
H_0: \mu_1 = \mu_2 = \dots = \mu_k \\
H_1: \text{al menos una media de condición difiere de las demás}
$$

> #### **Nota:** $H_1$ **no dice cuál** condición difiere. Para eso se necesita un **post-hoc** (pairwise_tests de pingouin, con corrección Holm o Bonferroni) una vez rechazada $H_0$.

## Supuestos

* #### 1. **Independencia entre sujetos**: las observaciones de sujetos distintos son independientes.
* #### 2. **Normalidad** (aproximada) de los **residuos** del modelo.
* #### 3. **Esfericidad**: las varianzas de las diferencias entre todos los pares de condiciones son iguales. `pingouin.rm_anova` calcula automáticamente la prueba de **Mauchly** y devuelve las columnas `sphericity`, `W_spher` y `p_spher`.

## ANOVA de medidas repetidas vs. Friedman

| Aspecto | ANOVA de medidas repetidas | Friedman |
| :--- | :--- | :--- |
| **Tipo** | Paramétrica | No paramétrica |
| **Qué compara** | Medias | Medianas (rangos) |
| **Supuestos** | Normalidad + esfericidad | Solo dependencia + ordinalidad |
| **Potencia** | Mayor si se cumplen supuestos | Menor, pero robusta |
| **Post-hoc** | `pingouin.pairwise_tests` | Nemenyi o Wilcoxon + Holm |
| **Cuándo usarlo** | Datos normales y esfericidad OK | Datos no normales u ordinales |

## Ejemplos en Machine Learning y PLN

| Situación | Por qué es de medidas repetidas |
| :--- | :--- |
| **Tres arquitecturas de Transformer** (BERT, RoBERTa, DistilBERT) evaluadas sobre **los mismos 10 datasets de clasificación de texto** | Cada dataset se evalúa con las 3 arquitecturas |
| **Tres estrategias de tokenización** (WordPiece, BPE, SentencePiece) aplicadas al **mismo corpus de reseñas** en 8 particiones de validación cruzada | Cada partición se tokeniza con las 3 estrategias |
| **Tres técnicas de aumento de datos** (back-translation, synonym replacement, random insertion) evaluadas sobre **los mismos 5 conjuntos de noticias** con 10 semillas cada uno | Cada combinación (conjunto, semilla) se evalúa con las 3 técnicas |
| **Tres modelos de embeddings** (Word2Vec, GloVe, FastText) usados como entrada para la **misma tarea de NER** sobre 6 corpus distintos | Cada corpus se entrena con los 3 embeddings |
| **Tres estrategias de fine-tuning** (full fine-tuning, LoRA, prefix-tuning) aplicadas a **los mismos 12 pares de frases** en una tarea de similitud semántica | Cada par se evalúa con las 3 estrategias |

## Parámetros principales

- #### **`data`**: DataFrame en formato **largo** (una fila por observación). Debe contener columnas para el sujeto, la condición y la variable dependiente.
- #### **`dv`**: nombre de la columna con la **variable dependiente** (la medición).
- #### **`within`**: nombre de la columna del **factor de medidas repetidas** (la condición).
- #### **`subject`**: nombre de la columna con el **identificador del sujeto** (unidad experimental).
- #### **`correction`**: `'auto'` (por defecto) calcula Mauchly y aplica Greenhouse-Geisser si es necesario; `True` fuerza la corrección; `False` no corrige.
- #### **`detailed`**: si es `True`, devuelve la tabla ANOVA completa.
- #### **`effsize`**: tipo de tamaño del efecto: `'ng2'` (eta cuadrado generalizado, por defecto).

## Valor de retorno

`pingouin.rm_anova` devuelve un DataFrame con las columnas [citation:2]:
- **`Source`**: nombre del factor de medidas repetidas.
- **`ddof1`**: grados de libertad del numerador.
- **`ddof2`**: grados de libertad del denominador.
- **`F`**: el estadístico $F$.
- **`p_unc`**: valor p sin corregir.
- **`ng2`**: tamaño del efecto (eta cuadrado generalizado).
- **`eps`**: factor épsilon de Greenhouse-Geisser.
- **`p_GG_corr`**: valor p corregido por Greenhouse-Geisser.
- **`W_spher`**: estadístico de la prueba de esfericidad de Mauchly.
- **`p_spher`**: valor p de la prueba de esfericidad.
- **`sphericity`**: booleano que indica si se cumple la esfericidad.

````python
import pandas as pd
import pingouin as pg

# Datos en formato largo: una fila por (desarrollador, lenguaje)
df_long = pd.DataFrame({
    'desarrollador': ['dev1', 'dev1', 'dev1',
                      'dev2', 'dev2', 'dev2',
                      'dev3', 'dev3', 'dev3',
                      'dev4', 'dev4', 'dev4',
                      'dev5', 'dev5', 'dev5'],
    'lenguaje':      ['Python', 'Java', 'Go',
                      'Python', 'Java', 'Go',
                      'Python', 'Java', 'Go',
                      'Python', 'Java', 'Go',
                      'Python', 'Java', 'Go'],
    'tiempo':        [45, 60, 50,
                      38, 55, 42,
                      52, 65, 58,
                      41, 58, 47,
                      48, 62, 53]
})
print(df_long)
````

> **Residuo = lo que queda después de quitar el efecto del sujeto y el efecto de la condición.**

## 2. La fórmula

Para cada observación $x_{ij}$ (sujeto $i$, condición $j$):

$$
\text{residuo}_{ij} = x_{ij} - \bar{x}_{i.} - \bar{x}_{.j} + \bar{x}
$$

donde:

| Símbolo | Significado | Cómo se calcula |
| :--- | :--- | :--- |
| $x_{ij}$ | Valor observado | El dato original |
| $\bar{x}_{i.}$ | Media del **sujeto** $i$ | Promedio de la fila $i$ (todas las condiciones de ese sujeto) |
| $\bar{x}_{.j}$ | Media de la **condición** $j$ | Promedio de la columna $j$ (todos los sujetos en esa condición) |
| $\bar{x}$ | Media **global** | Promedio de todos los datos |


````python
from scipy import stats

# 1. Calcular los residuos 
#    residuo_ij = x_ij - media_sujeto_i - media_condicion_j + media_global
media_global = df_long['tiempo'].mean()
df_long['media_sujeto'] = df_long.groupby('desarrollador')['tiempo'].transform('mean')
df_long['media_condicion'] = df_long.groupby('lenguaje')['tiempo'].transform('mean')
df_long['residuo'] = (df_long['tiempo'] 
                      - df_long['media_sujeto'] 
                      - df_long['media_condicion'] 
                      + media_global)

# 2. Aplicar Shapiro-Wilk sobre los residuos
W, p_shapiro = stats.shapiro(df_long['residuo'])
print(f"Shapiro-Wilk sobre residuos: W = {W:.4f}, p = {p_shapiro:.4f}")

if p_shapiro > 0.05:
    print("→ Normalidad de los residuos OK → ANOVA de medidas repetidas válido")
else:
    print("→ Normalidad de los residuos VIOLADA → considerar Friedman")
````

````python
# ============================================================
# ANOVA de medidas repetidas con pingouin
# ============================================================
# Se contrasta si el tiempo medio de implementación difiere
# entre los 3 lenguajes (Python, Java, Go), medidos sobre los
# MISMOS 5 desarrolladores. Diseño pareado de un factor.
#
# H0: mu_Python = mu_Java = mu_Go  (las 3 medias son iguales)
# H1: al menos una media difiere
# ============================================================

aov = pg.rm_anova(
    # DataFrame en formato LARGO: una fila por (sujeto, condición).
    # Debe contener las columnas: desarrollador, lenguaje, tiempo.
    data=df_long,

    # Variable dependiente: la medición que se compara entre condiciones.
    # En este caso, el tiempo de implementación (minutos).
    dv='tiempo',

    # Factor de medidas repetidas: la condición que se repite en cada sujeto.
    # Cada desarrollador se mide en los 3 lenguajes.
    within='lenguaje',

    # Identificador del sujeto (unidad experimental).
    # Se usa para separar la variabilidad entre sujetos del error.
    subject='desarrollador',

    # correction=True → calcula la prueba de esfericidad de Mauchly
    # y aplica la corrección de Greenhouse-Geisser si es necesario.
    # (Con 'auto' pingouin decide; con True fuerza el cálculo.)
    # El factor épsilon aplicado queda en la columna 'eps'.
    correction=True,

    # detailed=True → devuelve la tabla completa con dos filas:
    #   - la fila del factor ('lenguaje')
    #   - la fila del error ('Error' / 'Within')
    # Si fuera False, solo devuelve la fila del factor.
    detailed=True
)

# Imprime la tabla ANOVA completa con las columnas:
#   Source, SS, DF, MS, F, p_unc, p_GG_corr, ng2, eps,
#   sphericity, W_spher, p_spher
# (las últimas 4 solo aparecen si la versión de pingouin las expone)
print(aov)
````

| Columna | Valor | Interpretación |
| :--- | :---: | :--- |
| **SS** | 596.80 (lenguaje) / 11.87 (error) | La variabilidad explicada por el lenguaje es **50 veces mayor** que el error residual |
| **DF** | 2 (lenguaje) / 8 (error) | $k-1 = 2$ condiciones menos 1; $(k-1)(n-1) = 2 \times 4 = 8$ |
| **MS** | 298.40 (lenguaje) / 1.483 (error) | La varianza entre condiciones es **201 veces** la varianza residual |
| **F** | **201.17** | Estadístico de prueba. Bajo $H_0$, $F \approx 1$. Un valor de 201 es **extremadamente grande** |
| **p_unc** | $1.44 \times 10^{-7}$ | p-valor sin corregir. **Mucho menor que 0.05** |
| **p_GG_corr** | $4.3 \times 10^{-5}$ | p-valor corregido por Greenhouse-Geisser. **También $\ll 0.05$** |
| **ng2** | **0.646** | Tamaño del efecto: el lenguaje explica el **64.6 %** de la varianza. **Efecto grande** |
| **eps** | **0.586** | Épsilon de Greenhouse-Geisser. Como es **< 1**, hay violación de esfericidad y la corrección GG es la adecuada |
| **sphericity** | **True** | Mauchly no rechaza la esfericidad ($p = 0.159 > 0.05$). Aunque `eps < 1`, formalmente no se viola |
| **W_spher** | 0.2935 | Estadístico de Mauchly |
| **p_spher** | 0.1590 | p-valor de Mauchly. **> 0.05** → no se rechaza la esfericidad. $H_0$: cumple el supuesto de esfecidad (las varianzas de las diferencias entre todos los pares posibles de condiciones (o niveles del factor) son iguales). En este caso, esta prueba tiene muy poca evidencia, se recomienda considerar eps sobre Maucly. |

# Valores de épsilon (ε) y corrección por esfericidad

| Valor de ε | Interpretación | Corrección recomendada |
| :---: | :--- | :--- |
| **ε = 1** | Esfericidad perfecta. No se necesita corrección. | Ninguna |
| **ε < 1** | Violación de la esfericidad | **Greenhouse-Geisser (GG)** —  controla el Error Tipo I |

</br>
</br>

> ### **Nota:** el valor de `eps` que reporta `pingouin` es el épsilon de Greenhouse-Geisser. Si `eps < 1`, la corrección ya se aplicó y se debe reportar `p_GG_corr` (o el `p_unc` corregido internamente). Si `eps = 1`, la esfericidad es perfecta y se puede usar `p_unc`.


# Valores de η² (eta cuadrado) y su interpretación

## 1. La tabla de referencia

| Valor de η² | Interpretación |
| :---: | :--- |
| **η² < 0.01** | Efecto insignificante |
| **0.01 ≤ η² < 0.06** | Efecto pequeño |
| **0.06 ≤ η² < 0.14** | Efecto mediano |
| **η² ≥ 0.14** | Efecto grande |

</br>
</br>

> ### En el ejemplo, el **64.6 %** de la varianza en el tiempo de implementación se explica por el lenguaje de programación. Es un efecto **muy grande**, no solo estadísticamente significativo.



````python
import pandas as pd
import pingouin as pg

# ============================================================
# Extraer valores relevantes
# ============================================================
fila_factor = aov.iloc[0]
fila_error  = aov.iloc[1]

F_val   = fila_factor['F']
p_unc   = fila_factor['p_unc']
p_gg    = fila_factor.get('p_GG_corr', p_unc)
eta     = fila_factor['ng2']
eps     = fila_factor['eps']
gl_num  = fila_factor['DF']
gl_den  = fila_error['DF']

# Esfericidad explícita
spher = pg.sphericity(
    data=df_long, dv='tiempo', within='lenguaje', subject='desarrollador'
)

# ============================================================
# Tabla resumen (solo lo relevante)
# ============================================================
print("=" * 65)
print("RESUMEN DEL ANOVA DE MEDIDAS REPETIDAS")
print("=" * 65)
print(f"Factor:              lenguaje (Python, Java, Go)")
print(f"Sujetos:             {df_long['desarrollador'].nunique()} desarrolladores")
print(f"Observaciones:       {len(df_long)}")
print("-" * 65)
print(f"Estadístico:         F({gl_num}, {gl_den}) = {F_val:.4f}")
print(f"p-valor (sin corr.): {p_unc:.4e}")
print(f"p-valor (GG corr.):  {p_gg:.4e}")
print(f"Tamaño del efecto:   η²_G = {eta:.4f}", end="")
if eta < 0.01:
    print("  (insignificante)")
elif eta < 0.06:
    print("  (pequeño)")
elif eta < 0.14:
    print("  (mediano)")
else:
    print("  (grande)")
print("-" * 65)
print(f"Esfericidad (Mauchly): W = {spher.W:.4f}, p = {spher.pval:.4e}")
print(f"  → {'OK (no se viola)' if spher.spher else 'VIOLADA'}")
print(f"Épsilon (GG):        {eps:.4f}")
if eps < 1:
    print("  → Violación de esfericidad (usar Greenhouse-Geisser)")
print("-" * 65)

# Decisión
alpha = 0.05
p_final = p_unc if spher.spher else p_gg
print(f"DECISIÓN (α = {alpha}):")
if p_final < alpha:
    print(f"  → Se RECHAZA H0 (p = {p_final:.4e} < {alpha})")
    print("  → Al menos una condición tiene una media distinta")
    print("  → Siguiente paso: post-hoc (pairwise_tests + Holm)")
else:
    print(f"  → NO se rechaza H0 (p = {p_final:.4e} >= {alpha})")
    print("  → No hay evidencia de diferencias entre condiciones")
print("=" * 65)
````

# Prueba posthoc pairwise_tests

````python
import pandas as pd
import pingouin as pg

# ============================================================
# Post-hoc para ANOVA de medidas repetidas
# ============================================================
# Se ejecuta SOLO si el ANOVA global rechazó H0.
# Compara todos los pares de condiciones con corrección Holm.
#
# Hipótesis por par:
#   H0: mu_i = mu_j  (las dos condiciones tienen la misma media)
#   H1: mu_i != mu_j (las medias difieren)
# ============================================================

posthoc = pg.pairwise_tests(
    data=df_long,             # DataFrame en formato largo (una fila por medición)
    dv='tiempo',              # Variable dependiente numérica que se compara (minutos)
    within='lenguaje',        # Factor intrasujeto cuyas categorías se comparan par a par
    subject='desarrollador',  # Columna que identifica cada bloque/sujeto (activa el modo pareado)
    padjust='holm',           # Método de Holm-Bonferroni para ajustar p-valores (columna 'p-corr')
    effsize='hedges'          # Calcula la g de Hedges para medir la magnitud práctica de la diferencia
)


print("=" * 90)
print("POST-HOC: pairwise_tests (medidas repetidas) + corrección Holm")
print("=" * 90)
print(posthoc[['Contrast', 'A', 'B', 'T', 'dof', 'p_unc', 'p_corr', 'hedges']]
      .to_string(index=False))
print()

# ============================================================
# Interpretación automática
# ============================================================
alpha = 0.05
print("=" * 90)
print("INTERPRETACIÓN")
print("=" * 90)

for _, fila in posthoc.iterrows():
    a = fila['A']
    b = fila['B']
    p_aj = fila['p_corr']
    t_val = fila['T']
    d = fila['hedges']

    if p_aj < alpha:
        # Determinar dirección usando las medias
        media_a = df_long[df_long['lenguaje'] == a]['tiempo'].mean()
        media_b = df_long[df_long['lenguaje'] == b]['tiempo'].mean()
        if media_a < media_b:
            direccion = f"{a} es más rápido que {b}"
            print(f"  {a} vs {b}: p_aj = {p_aj:.4e} → DIFERENCIA SIGNIFICATIVA")
            print(f"    → {direccion} "
                f"({media_a:.1f} vs {media_b:.1f} min)")

        else:
            direccion = f"{b} es más rápido que {a}"
            print(f"  {b} vs {a}: p_aj = {p_aj:.4e} → DIFERENCIA SIGNIFICATIVA")
            print(f"    → {direccion} "
                f"({media_b:.1f} vs {media_a:.1f} min)")
        print(f"    T = {t_val:.4f}, Hedges' g = {d:.4f}")
    else:
        print(f"  {a} vs {b}: p_aj = {p_aj:.4e} → NO significativo")
    print()
````

# `T` y `hedges` en el post-hoc

## 1. `T` — Estadístico t de la prueba pareada

### Interpretación

- ### **$|T|$ grande** → la diferencia media es grande **en relación con la variabilidad** de las diferencias → evidencia contra $H_0$.
- ### **$T \approx 0$** → no hay diferencia consistente entre las condiciones.
- ### **Signo de $T$** → indica la **dirección** de la diferencia (qué condición es mayor).

> ###  **Ejemplo con los datos:** `T = -14.23` para Python vs Java significa que Python tiene un tiempo **menor** que Java (el signo negativo indica que $A - B < 0$, es decir, Python < Java).

## 2. `hedges` — Tamaño del efecto (Hedges' g)

### `hedges` es el **tamaño del efecto** ( **diferencia real entre las condiciones**) medido con la **g de Hedges**.


### Interpretación

| $\|g\|$ | Interpretación |
| :---: | :--- |
| 0.2 | Efecto pequeño |
| 0.5 | Efecto mediano |
| 0.8 | Efecto grande |
| **> 2** | **Efecto enorme** |


### Signo de `hedges`

- ### **Positivo** → la media de $A$ es **mayor** que la de $B$.
- ### **Negativo** → la media de $A$ es **menor** que la de $B$.

> ### **Ejemplo:** `hedges = -3.84` para Python vs Java significa que Python tiene un tiempo **mucho menor** que Java, y la magnitud (3.84) indica un **efecto enorme**.


# Prueba de Friedman

### `scipy.stats.friedmanchisquare(*samples, axis=0, nan_policy='propagate', keepdims=False)`

* #### Contrasta si **tres o más condiciones pareadas** (medidas repetidas sobre la misma unidad experimental) provienen de la **misma distribución**.
* #### Es la alternativa **no paramétrica** al ANOVA de medidas repetidas. Se usa cuando **no se cumple el supuesto de normalidad** de los residuos o cuando los datos son al menos **ordinales**.
* #### A diferencia del ANOVA de medidas repetidas, **no requiere esfericidad** ni normalidad. Solo requiere que las observaciones sean **dependientes** (los mismos sujetos en todas las condiciones) y que los datos puedan **ordenarse**.


## ¿En qué consiste la prueba de Friedman?

* #### La prueba de Friedman **no compara las condiciones par a par**, sino que contrasta **globalmente** si todas las condiciones provienen de la misma distribución.
* #### Procedimiento:
  1. #### **Organizar los datos en bloques**: cada fila es un sujeto (bloque), cada columna es una condición.
  2. #### **Rankear dentro de cada bloque**: para cada sujeto, se ordenan sus mediciones de menor a mayor y se asignan rangos $1, 2, \dots, k$ (donde $k$ es el número de condiciones). En caso de empates, se asigna el rango promedio.
  3. #### **Sumar los rangos por condición**: $R_j = \sum_{i=1}^{n} R_{ij}$, donde $R_{ij}$ es el rango del sujeto $i$ en la condición $j$.
  4. #### **Calcular el estadístico de Friedman**.
* #### La lógica es similar a Kruskal-Wallis, pero **bloqueando por sujeto**.

## 1. Fórmulas

### Sea:
- #### $k$ = número de condiciones.
- #### $n$ = número de sujetos (bloques).
- #### $R_{ij}$ = rango de la observación del sujeto $i$ en la condición $j$.
- #### $R_j = \sum_{i=1}^{n} R_{ij}$ = suma de rangos de la condición $j$.

### 1.1 Estadístico de Friedman (sin empates)

$$
\large
\chi^2_F = \frac{12}{n k (k+1)} \sum_{j=1}^{k} R_j^2 - 3n(k+1)
$$

#### donde $R_j$ es la suma de rangos de la condición $j$.

### 1.2 Estadístico de Friedman (con corrección por empates)

Cuando hay empates, el estadístico se corrige:

$$
\large
\chi^2_F = \frac{12}{n k (k+1)} \sum_{j=1}^{k} R_j^2 - 3n(k+1) \Big/ C
$$

#### donde $C$ es un factor de corrección por empates. SciPy aplica esta corrección automáticamente.

### 1.3 Grados de libertad y valor p

- #### Bajo $H_0$, el estadístico $\chi^2_F$ sigue aproximadamente una distribución **$\chi^2$ (chi-cuadrado)** con $k - 1$ grados de libertad [citation:1][citation:2].
- #### **Limitación importante**: la aproximación $\chi^2$ es fiable solo si $n > 10$ y hay más de 6 muestras repetidas [citation:4]. Con pocos sujetos, el valor p puede ser impreciso.

## Las hipótesis son:

$$
\large
H_0: \text{las } k \text{ condiciones tienen la misma distribución (medianas iguales)} \\
H_1: \text{al menos una condición tiene una distribución diferente}
$$

> #### **Nota:** $H_1$ **no dice cuál** condición difiere. Para eso se necesita un **post-hoc** (Nemenyi o Wilcoxon con corrección Holm) una vez rechazada $H_0$.

## Supuestos

* #### 1. **Independencia entre bloques**: las observaciones de sujetos distintos son independientes.
* #### 2. **Datos al menos ordinales**: las mediciones deben poder ordenarse.
* #### 3. **Mínimo de sujetos**: se recomienda al menos **5 bloques** para que la aproximación $\chi^2$ sea válida [citation:1]. Con menos, el valor p puede ser impreciso.
* #### 4. **No requiere normalidad ni esfericidad**.

## Friedman vs. ANOVA de medidas repetidas

| Aspecto | Friedman | ANOVA de medidas repetidas |
| :--- | :--- | :--- |
| **Tipo** | No paramétrica | Paramétrica |
| **Qué compara** | Medianas (rangos) | Medias |
| **Supuestos** | Solo dependencia + ordinalidad | Normalidad + esfericidad |
| **Potencia** | Menor, pero robusta | Mayor si se cumplen supuestos |
| **Post-hoc** | Nemenyi o Wilcoxon + Holm | `pairwise_tests` (pingouin) |
| **Cuándo usarlo** | Datos no normales u ordinales | Datos normales y esfericidad OK |

## Ejemplos

| Situación | Por qué es de medidas repetidas |
| :--- | :--- |
| **Tres lenguajes** implementados por **los mismos desarrolladores** | Cada desarrollador se mide en los 3 lenguajes |
| **Tres arquitecturas de Transformer** evaluadas sobre **los mismos 10 datasets** | Cada dataset se evalúa con las 3 arquitecturas |
| **Tres herramientas** aplicadas a **los mismos proyectos** | Cada proyecto se mide con las 3 herramientas |
| **Tres metodologías** probadas por **los mismos equipos** | Cada equipo se mide en las 3 metodologías |

## Parámetros principales

- #### **`*samples`**: tres o más arrays (uno por condición). Todos deben tener el **mismo número de elementos** (cada posición $i$ corresponde al mismo sujeto) [citation:4].
- #### **`axis`**: eje sobre el que se calcula la prueba (por defecto `0`).
- #### **`nan_policy`**: `'propagate'` (por defecto), `'omit'` o `'raise'`.
- #### **`keepdims`**: si es `True`, conserva las dimensiones reducidas.

## Valor de retorno

#### Devuelve un objeto `FriedmanchisquareResult` con:
- #### **`statistic`**: el estadístico $\chi^2_F$ (corregido por empates).
- #### **`pvalue`**: el valor p asociado a $H_0$, basado en la distribución $\chi^2$ con $k-1$ grados de libertad [citation:4].

## Post-hoc: Nemenyi (o Wilcoxon + Holm)

Si Friedman rechaza $H_0$ ($p < 0.05$), se necesita un **post-hoc** para identificar **qué pares de condiciones difieren**.

### Opción A — Nemenyi (recomendada para Friedman)

La prueba de **Nemenyi** compara todos los pares de condiciones controlando el error tipo I. Está disponible en `scikit-posthocs` [citation:5].

```python
import scikit_posthocs as sp
import pandas as pd

# Preparar datos en formato largo
df_long = pd.DataFrame({
    'sujeto':     ['dev1', 'dev1', 'dev1', 'dev2', 'dev2', 'dev2', ...],
    'lenguaje':   ['Python', 'Java', 'Go', 'Python', 'Java', 'Go', ...],
    'tiempo':     [45, 60, 50, 38, 55, 42, ...]
})

# Nemenyi para Friedman
nemenyi = sp.posthoc_nemenyi_friedman(
    df_long,
    y_col='tiempo',
    block_col='sujeto',
    group_col='lenguaje',
    melted=True
)
print(nemenyi)

````python
import pandas as pd
import numpy as np
from scipy.stats import friedmanchisquare
import scikit_posthocs as sp

# ============================================================
# Datos en formato largo: una fila por (desarrollador, lenguaje)
# n = 10 desarrolladores × 3 lenguajes (medidas repetidas)
# ============================================================
df_long = pd.DataFrame({
    'desarrollador': ['dev1', 'dev1', 'dev1',
                      'dev2', 'dev2', 'dev2',
                      'dev3', 'dev3', 'dev3',
                      'dev4', 'dev4', 'dev4',
                      'dev5', 'dev5', 'dev5',
                      'dev6', 'dev6', 'dev6',
                      'dev7', 'dev7', 'dev7',
                      'dev8', 'dev8', 'dev8',
                      'dev9', 'dev9', 'dev9',
                      'dev10', 'dev10', 'dev10'],
    'lenguaje':      ['Python', 'Java', 'Go'] * 10,
    'tiempo':        [45, 60, 50,
                      38, 55, 42,
                      52, 65, 58,
                      41, 58, 47,
                      48, 62, 53,
                      50, 63, 55,
                      43, 57, 46,
                      47, 61, 52,
                      39, 56, 44,
                      51, 120, 57]
})

print("Primeras filas del DataFrame:")
print(df_long.head(9))
print(f"\nDimensiones: {df_long.shape}")
print(f"Desarrolladores: {df_long['desarrollador'].nunique()}")
print(f"Lenguajes: {df_long['lenguaje'].unique().tolist()}")
````

````python
# ============================================================
# Verificación de supuestos con pandas
# (normalidad de residuos + esfericidad)
# ============================================================
# Para el ANOVA de medidas repetidas se requieren DOS supuestos:
#   1. Normalidad de los RESIDUOS del modelo (Shapiro-Wilk).
#   2. Esfericidad de las diferencias entre pares de condiciones
#      (prueba de Mauchly).
#
# Si la normalidad falla → usar Friedman (no paramétrica).
# Si la esfericidad falla → aplicar corrección Greenhouse-Geisser
#   (o Huynh-Feldt si ε > 0.75).
# ============================================================

import pandas as pd
import numpy as np
from scipy import stats
import pingouin as pg

# ============================================================
# 1. NORMALIDAD DE LOS RESIDUOS (Shapiro-Wilk) — con pandas
# ============================================================
# residuo_ij = x_ij - media_sujeto_i - media_condicion_j + media_global
#
# Con pandas, las medias se calculan con groupby + transform,
# que devuelve una Serie alineada con el índice original.

# Media global
media_global = df_long['tiempo'].mean()

# Media por sujeto (se repite para cada fila del mismo sujeto)
df_long['media_sujeto'] = (
    df_long.groupby('desarrollador')['tiempo']
    .transform('mean')
)

# Media por condición (se repite para cada fila del mismo lenguaje)
df_long['media_condicion'] = (
    df_long.groupby('lenguaje')['tiempo']
    .transform('mean')
)

# Residuo
df_long['residuo'] = (
    df_long['tiempo']
    - df_long['media_sujeto']
    - df_long['media_condicion']
    + media_global
)

# Shapiro-Wilk sobre los residuos
W, p_shapiro = stats.shapiro(df_long['residuo'])

print("=" * 60)
print("1. NORMALIDAD DE LOS RESIDUOS (Shapiro-Wilk)")
print("=" * 60)
print(f"   W = {W:.4f}")
print(f"   p = {p_shapiro:.4e}")
if p_shapiro > 0.05:
    print("   → No se rechaza H0: residuos compatibles con normalidad")
    normalidad_ok = True
else:
    print("   → Se RECHAZA H0: residuos NO normales")
    print("   → Considerar Friedman como alternativa no paramétrica")
    normalidad_ok = False


# ============================================================
# 2. ESFERICIDAD (Mauchly) — con pandas
# ============================================================
# La esfericidad se calcula sobre la matriz sujetos × condiciones.
# Con pandas: pivotar el DataFrame largo a ancho y pasar el ancho
# a pg.sphericity (o a pg.rm_anova con correction=True).

df_wide =  df_long.pivot(index='desarrollador', columns='lenguaje', values='tiempo')

# Reordenar las columnas
df_wide = df_wide[['Python', 'Java', 'Go']]

print("\nMatriz sujetos x condiciones:")
print(df_wide)

# pg.sphericity directo sobre el DataFrame ancho
# (pingouin acepta un DataFrame ancho si se le pasan las columnas)
spher = pg.sphericity(df_wide)

print("\n" + "=" * 60)
print("2. ESFERICIDAD (Mauchly)")
print("=" * 60)
print(f"   W de Mauchly = {spher.W:.4f}")
print(f"   p-valor      = {spher.pval:.4e}")
print(f"   ¿Esférico?   = {spher.spher}")

if spher.spher:
    print("   → No se rechaza H0: esfericidad asumida")
    esfericidad_ok = True
else:
    print("   → Se RECHAZA H0: esfericidad VIOLADA")
    print("   → Aplicar corrección Greenhouse-Geisser")
    esfericidad_ok = False


# ============================================================
# 3. RESUMEN Y DECISIÓN
# ============================================================
print("\n" + "=" * 60)
print("3. RESUMEN DE SUPUESTOS")
print("=" * 60)
print(f"   Normalidad de residuos: {'OK' if normalidad_ok else ' NO'}")
print(f"   Esfericidad:            {'OK' if esfericidad_ok else ' NO'}")

print("\n   Decisión:")
if not normalidad_ok:
    print("   → Usar FRIEDMAN (no paramétrica)")
elif not esfericidad_ok:
    print("   → Usar ANOVA de medidas repetidas CON corrección GG")
else:
    print("   → Usar ANOVA de medidas repetidas estándar")
````

````python
# ============================================================
# Prueba de Friedman
# ============================================================
# H0: las 3 condiciones tienen la misma distribución (medianas iguales)
# H1: al menos una condición tiene una distribución diferente
#
# Friedman NO requiere normalidad ni esfericidad.
# Solo requiere dependencia entre condiciones y datos ordinales.

python  = df_wide['Python'].values
java = df_wide['Java'].values
go = df_wide['Go'].values


stat, p = friedmanchisquare(python, java, go)
print("=" * 55)
print("PRUEBA DE FRIEDMAN")
print("=" * 55)
print(f"chi2_F = {stat:.4f}")
print(f"p-valor = {p:.4e}")

alpha = 0.05
if p < alpha:
    print(f"→ Se RECHAZA H0 (p = {p:.4e} < {alpha})")
    print("→ Al menos un lenguaje tiene una distribución distinta")
    rechaza = True
else:
    print(f"→ NO se rechaza H0 (p = {p:.4e} >= {alpha})")
    print("→ No hay evidencia de diferencias entre lenguajes")
    rechaza = False
````

````python
# ============================================================
# Post-hoc: Nemenyi (solo si se rechaza H0)
# ============================================================
if rechaza:
    print("\n" + "=" * 55)
    print("POST-HOC NEMENYI")
    print("=" * 55)


    # Nemenyi sobre la matriz ancha
    nemenyi = sp.posthoc_nemenyi_friedman(df_wide)
    print("P-valores ajustados (Nemenyi):")
    print(nemenyi)

    # Interpretación automática
    print("\nInterpretación:")
    lenguajes = ['Python', 'Java', 'Go']
    for i, a in enumerate(lenguajes):
        for j, b in enumerate(lenguajes):
            if i < j:
                p_aj = nemenyi.loc[a, b]
                if p_aj < 0.05:
                    print(f"  {a} vs {b}: p_aj = {p_aj:.4f} → DIFERENCIA SIGNIFICATIVA")
                else:
                    print(f"  {a} vs {b}: p_aj = {p_aj:.4f} → no significativo")
else:
    print("\nNo se rechaza H0: no se ejecuta post-hoc.")
````

````python
# ============================================================
# Resumen descriptivo por lenguaje: mediana, cuartiles, IQR,
# mínimo, máximo y número de observaciones
# ============================================================
# Este resumen es para acompañar la prueba de Friedman,
# porque Friedman contrasta MEDIANAS (rangos).
#
# Qué incluye cada columna:
#   - mediana: medida de tendencia central coherente con Friedman.
#   - q1, q3: primer y tercer cuartil (posición del 50 % central).
#   - iqr: rango intercuartílico (q3 - q1), medida de dispersión
#          robusta a outliers.
#   - min, max: extremos observados (útiles para detectar outliers).
#   - n: número de observaciones por lenguaje (10 en cada uno).
#
# El .reindex(['Python', 'Java', 'Go']) fuerza el orden de las filas
# para que el resumen coincida con el orden usado en el post-hoc
# y en la conclusión.

resumen = (
    df_long.groupby('lenguaje')['tiempo']
    .agg(                                                  # permite aplicar una o varias funciones de agregación a un grupo de datos, devolviendo un resumen compacto
        mediana='median',                                  # mediana
        media='mean',                                      # media 
        std='std',                                         # desviación estándar 
        q1=lambda x: x.quantile(0.25),                     # primer cuartil
        q3=lambda x: x.quantile(0.75),                     # tercer cuartil
        iqr=lambda x: x.quantile(0.75) - x.quantile(0.25), # rango intercuartílico
        min='min',                                         # mínimo
        max='max',                                         # máximo
        n='count'                                          # número de observaciones
    )
    .reindex(['Python', 'Java', 'Go'])                     # orden explícito
)

print(resumen)
````
