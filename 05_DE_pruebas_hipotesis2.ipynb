## ¿Qué son las pruebas de hipótesis?

### Son procedimientos estadísticos que nos permiten **tomar decisiones** sobre una población a partir de los datos de una muestra. Nos ayudan a responder preguntas como:

> ### *"¿La nueva herramienta reduce realmente el tiempo de desarrollo, o la diferencia observada es solo producto del azar?"*

## Objetivos

### 1. Entender qué es una prueba de hipótesis.
### 2. Diferenciar las pruebas paramétricas de las no paramétricas.
### 3. Identificar cuándo usar cada una según los supuestos de los datos.
### 4. Conocer las pruebas más comunes de cada familia.

# 1. La decisión fundamental: usar una prueba ¿paramétrica o no paramétrica?

### Antes de aplicar cualquier prueba, debemos hacernos una pregunta clave: **¿mis datos cumplen los supuestos?**

| <span style="font-size: 28px;">**Tipo de prueba**</span> | <span style="font-size: 28px;">**¿Qué asume?**</span> | <span style="font-size: 28px;">**¿Qué usa?**</span> | <span style="font-size: 28px;">**¿Cuándo usarla?**</span> |
| :--- | :--- | :--- | :--- |
| <span style="font-size: 28px;"> **Paramétrica**</span> | <span style="font-size: 28px;"> Normalidad y (usualmente) homocedasticidad.</span>  | <span style="font-size: 28px;">Medias y varianzas. </span>| <span style="font-size: 28px;">Cuando los datos cumplen los supuestos.</span> |
| <span style="font-size: 28px;">**No paramétrica**</span> | <span style="font-size: 28px;">No asume una distribución específica. | <span style="font-size: 28px;">Rangos u ordenaciones.</span> | <span style="font-size: 28px;">Cuando los datos no son normales, son ordinales o hay valores atípicos.</span> |

</br>

> ### **Importante:** Si los datos cumplen los supuestos, usar la prueba paramétrica (es más potente). Si no, usa la no paramétrica (es más segura).

## 2. El Mapa de las Pruebas de Hipótesis

### La siguiente tabla resume las pruebas más comunes según el **objetivo**, el **número de grupos** y el **tipo de diseño** (independiente o pareado):

| **Objetivo del análisis** | **Número de grupos** | **Prueba Paramétrica** | **Prueba No Paramétrica** |
| :--- | :---: | :--- | :--- |
| **Comparar medias** | 1 (vs. referencia) | Prueba $t$ de una muestra | Prueba de los signos / Wilcoxon |
| **Comparar medias** | 2 (independientes) | Prueba $t$ de Student (o Welch) | **U de Mann-Whitney** |
| **Comparar medias** | 2 (pareados) | Prueba $t$ pareada | **Wilcoxon** |
| **Comparar medias** | 3+ (independientes) | **ANOVA** (o ANOVA de Welch) | **Kruskal-Wallis** |
| **Comparar medias** | 3+ (pareados) | ANOVA de medidas repetidas | **Friedman** |
| **Comparar medias** | 3+ (2 factores) | ANOVA factorial (2 vías) | Alineación de rangos (ART) |
| **Analizar relación** | 2 variables continuas | Correlación de **Pearson** | Correlación de **Spearman** |
| **Analizar asociación** | 2 variables categóricas | Prueba $\chi^2$ | Prueba exacta de Fisher |

</br>
</br>

> **Importante:** 
> #### * Si los datos cumplen los supuestos (normalidad y homocedasticidad), usar paramétricas. 
> #### * Si no los cumplen, usar no paramétricas.
> #### * Si las variables son categóricas, usar $\chi^2$.

# Prueba t pareada

### `scipy.stats.ttest_rel(a, b,  alternative='two-sided')`

* ### Contrasta si la **media de las diferencias** entre dos muestras pareadas es cero. 

## ¿Qué significa "pareado"?

* ### **Pareado** (o de **muestras dependientes / relacionadas**) 
* ### Significa que cada observación del primer grupo está **vinculada de forma natural** con una observación del segundo grupo
* ### No son dos conjuntos de individuos distintos, sino **la misma unidad experimental medida dos veces** 
* ### Dos unidades emparejadas deliberadamente por compartir alguna característica.
* ### Se usa cuando las mediciones provienen de **los mismos sujetos/unidades** en dos condiciones distintas (antes/después, método A/método B sobre los mismos casos, mismas particiones en validación cruzada, etc.).


## ¿En qué consiste la prueba pareada?

* ### La prueba t pareada **no compara los dos grupos como si fueran independientes**, sino que trabaja **par a par**
 * ### Para cada unidad experimental (el mismo paciente, la misma muestra, el mismo fold) se calcula la **diferencia** entre las dos condiciones:

$$
\large
d_i = a_i - b_i \quad \text{para cada par } i = 1, \dots, n
$$

* ### A partir de esas $n$ diferencias, la prueba contrasta si su **media** es cero:

    $
    \large
    H_0: \mu_d = 0 \\
    H_1: \mu_d \neq 0 \ \text{(o } >0 \text{ / } <0 \text{ según } \textit{alternative}\text{)}
    $

### El estadístico es:

$$
\large
t = \frac{\bar{d}}{s_d / \sqrt{n}}
$$

### donde 
* ### $\bar{d}$ es la media de las diferencias
* ### $s_d$ su desviación estándar
* ### $n$ el número de pares. 
* ### Bajo $H_0$, este estadístico sigue una distribución $t$ de Student con $n-1$ grados de libertad.


### ¿Por qué es más potente que una prueba independiente?

* ### Porque al restar **dentro de cada par**, se **elimina la variabilidad entre unidades**. 
* ### Si el paciente 1 tiene valores altos en ambas condiciones y el paciente 2 tiene valores bajos en ambas, esa diferencia individual desaparece al calcular $d_i$. 
* ### Solo queda la **variabilidad del efecto** (lo que cambia entre condiciones), lo que hace más fácil detectar un efecto real.

> * ### **Requisito clave:** el emparejamiento debe ser **correcto**. Cada par debe compartir la fuente de variabilidad que se quiere controlar (el mismo sujeto, el mismo fold, el mismo dataset).

## Supuestos

* ### 1. **Observaciones pareadas**: cada $a_i$ está vinculado a un $b_i$ de la misma unidad.
* ### 2. **Independencia entre pares**: los pares no se influyen entre sí (folds disjuntos, sujetos distintos).
* ### 3. **Normalidad de las diferencias** $d_i$ (no de cada muestra por separado). Se verifica con **Shapiro-Wilk** sobre $d$. Si falla, usar **Wilcoxon** como alternativa no paramétrica.



### Ejemplos

| Situación | Por qué es pareado |
| :--- | :--- |
| **Antes / después** de un tratamiento en los mismos pacientes | El mismo paciente se mide dos veces (presión arterial antes del medicamento y después)|
| **Método A vs. Método B** sobre las mismas muestras | Cada muestra se procesa con ambos métodos |
| **Mismos folds de validación cruzada** con y sin estandarización con `StandardScaler` | Cada partición de datos genera un par de mediciones sobre el mismo división |
| **Preprocesamiento** Conversión a minúsculas y sin conversión al mismo texto | Cada partición de datos genera un par de mediciones sobre el mismo división |


## Parámetros principales

- #### **`a`**: primer conjunto de mediciones (array).
- #### **`b`**: segundo conjunto de mediciones (array), del mismo tamaño que `a`.
- #### **`alternative`**: dirección de la hipótesis alternativa:
  - #### **`'two-sided'`** (predeterminado): $\mu_d \neq 0$.
  - #### **`'greater'`**: $\mu_d > 0$ (la media de `a - b` es mayor que cero).
  - #### **`'less'`**: $\mu_d < 0$ (la media de `a - b` es menor que cero).

    ## Valores de `alternative`

    | `alternative` | $H_0$ | $H_1$ |
    | :--- | :--- | :--- |
    | **`'two-sided'`** (por defecto) | $\mu_d = 0$ | $\mu_d \neq 0$ |
    | **`'greater'`** | $\mu_d \leq 0$ | $\mu_d > 0$ (A tiende a ser mayor que B) |
    | **`'less'`** | $\mu_d \geq 0$ | $\mu_d < 0$ (A tiende a ser menor que B) |

    </br>
    </br>


## Valor de retorno

#### Devuelve :
- #### **`statistic`**: el estadístico $t$.
- #### **`pvalue`**: el valor p.
- #### **`df`**: grados de libertad.


## Ejemplo de uso

````python
import numpy as np
from scipy import stats

# Accuracy por partición (fold) (10-fold CV) sobre el mismo dataset
# Modelo SIN estandarización
sin_scaler = np.array([
    0.812, 0.798, 0.823, 0.805, 0.817,
    0.799, 0.810, 0.821, 0.808, 0.815
])

# Modelo CON StandardScaler (mismos folds)
con_scaler = np.array([
    0.848, 0.835, 0.851, 0.842, 0.856,
    0.839, 0.847, 0.853, 0.844, 0.850
])

# 1) Diferencias pareadas
d = con_scaler - sin_scaler
print(f"Media dif = {d.mean():.5f} | SD dif = {d.std(ddof=1):.5f}")

# 2) Supuesto de normalidad sobre las diferencias
shapiro_stat, shapiro_p = stats.shapiro(d)
print(f"Shapiro-Wilk: W={shapiro_stat:.4f}, p={shapiro_p:.4f}")

# 3) Prueba t pareada unilateral (H1: con_scaler > sin_scaler)
t_stat, p_val = stats.ttest_rel(con_scaler, sin_scaler, alternative='greater')
print(f"t = {t_stat:.4f}, p-valor = {p_val:.6e}")

````

# Interpretación del ejemplo: prueba t pareada con `StandardScaler`

## 1. Contexto del experimento

### Se compararon dos configuraciones del mismo modelo sobre **los mismos 10 folds de validación cruzada**:

- ### **`sin_scaler`**: modelo entrenado con las features en su escala original.
- ### **`con_scaler`**: mismo modelo, pero con las features estandarizadas (`StandardScaler`).

### Como cada fold genera un par de mediciones (con y sin scaler sobre el mismo split), el diseño es **pareado** → la prueba correcta es **`ttest_rel`**.

## 2. Hipótesis planteadas

### Con `alternative='greater'`:

### $$H_0: \mu_d \leq 0 \quad \text{(el scaler no mejora la accuracy media)}$$

### $$H_1: \mu_d > 0 \quad \text{(el scaler sí mejora la accuracy media)}$$

### donde $\mu_d = \mu_{\text{con\_scaler}} - \mu_{\text{sin\_scaler}}$.

## 3. Resultados y su lectura

### 3.1 Diferencias pareadas

### Media dif = 0.03570 | SD dif = 0.00347


- #### La accuracy media **aumenta en +0.0357** (≈ **+3.57 puntos porcentuales**) al usar `StandardScaler`.
- #### La desviación estándar de esas diferencias es **muy pequeña** (0.0035), lo que indica que la mejora es **consistente en los 10 folds**: no hay folds donde el scaler empeore.

### 3.2 Supuesto de normalidad de las diferencias

### Shapiro-Wilk: W=0.8846, p=0.1474

- ### $p = 0.1474 > 0.05$ → **no se rechaza la normalidad** de las diferencias.
- ### Se cumple el supuesto de `ttest_rel` → la prueba paramétrica es válida.
- ### (Si hubiera dado $p < 0.05$, habría que usar `wilcoxon` como alternativa no paramétrica.)

### 3.3 Prueba t pareada

- ### $t(9) = 32.57$ con $p = 5.96 \times 10^{-11}$.
- ### $p <  0.05$ → **se rechaza $H_0$** con $\alpha = 0.05$.
- ### Conclusión: **el `StandardScaler` mejora significativamente la accuracy media**.


# Prueba de Wilcoxon (alternativa no paramétrica a la prueba t pareada)

### `scipy.stats.wilcoxon(x, y=None,  alternative='two-sided')`

### Contrasta si la distribución de las diferencias `x - y` es simétrica respecto a cero. Es la versión no paramétrica de la prueba t pareada: se usa cuando **no se cumple el supuesto de normalidad de las diferencias**.

## Parámetros principales

- ####  **`x`**: primer conjunto de mediciones, o directamente las diferencias `x - y` si se omite `y`. Debe ser un array unidimensional.

- #### **`y`**: segundo conjunto de mediciones (opcional). Si se omite, `x` se interpreta como las diferencias. **Advertencia**: cuando se pasan `x` e `y` por separado, los errores de redondeo en la resta pueden asignar rangos distintos a diferencias que deberían ser iguales. Se recomienda calcular primero `d = x - y`, redondear si es necesario, y pasar solo `x=d`.


- #### **`alternative`**: define la hipótesis alternativa:
  - #### **`'two-sided'`** (predeterminado): la distribución de `d` **no es simétrica** respecto a cero.
  - #### **`'greater'`**: la distribución de `d` es estocásticamente **mayor** que una distribución simétrica respecto a cero (unilateral derecha).
  - #### **`'less'`**: la distribución de `d` es estocásticamente **menor** que una distribución simétrica respecto a cero (unilateral izquierda).

- #### **`method`**: método para calcular el valor p: `'auto'` (predeterminado), `'exact'`, `'asymptotic'` o una instancia de `PermutationMethod`.

### Valor de retorno

- #### **`statistic`**: si `alternative='two-sided'`, es la menor de las sumas de rangos por encima o por debajo de cero; en otro caso, la suma de rangos de las diferencias positivas.
- #### **`pvalue`**: el valor p calculado según `alternative` y `method`.
- #### **`zstatistic`**: el estadístico z normalizado (solo cuando `method='asymptotic'`).



---
## Idea sintética de los rangos en Wilcoxon

> #### **Wilcoxon no mira "cuánto" mejora cada par, sino "en cuántos pares" y "con qué intensidad relativa" gana una condición.**

#### El truco de los rangos:

1. #### **Se resta par a par**: $d_i = x_i - y_i$. Solo importa el **signo** (¿quién gana?) y el **orden** de las magnitudes.
2. #### **Se ordenan las diferencias por su valor absoluto** (de la más pequeña a la más grande) y se les asigna un **rango** $1, 2, \dots, n'$. La diferencia más pequeña recibe rango 1; la más grande, el rango máximo.
3. #### **Se separan los rangos por signo**: los rangos de las diferencias positivas van a $W^+$, los de las negativas a $W^-$.

## La intuición

> #### **Si una condición gana consistentemente, se lleva los rangos altos; la otra se queda con los bajos.** Si ambas ganan por igual, los rangos se reparten y $W^+ \approx W^-$.

## Ejemplo mini (4 pares)

| Par | $d_i$ | $\|d_i\|$ | Rango | Signo |
| :---: | :---: | :---: | :---: | :---: |
| 1 | +0.001 | 0.001 | **1** | + → $W^+$ |
| 2 | +0.010 | 0.010 | **3** | + → $W^+$ |
| 3 | −0.005 | 0.005 | **2** | − → $W^-$ |
| 4 | +0.020 | 0.020 | **4** | + → $W^+$ |

$$W^+ = 1 + 3 + 4 = 8, \qquad W^- = 2$$

- #### La condición "+" gana **3 de 4 pares** y además se lleva **los rangos más altos** (3 y 4) → $W^+$ grande.
- #### La condición "−" gana solo un par pequeño → $W^-$ chico.
- #### **Conclusión:** la condición "+" tiende a ser mayor.

## Por qué rangos y no promedios

- #### **Robusto a outliers**: una diferencia gigante recibe el rango máximo (no infla la suma como haría una media).
- #### **Solo usa el orden**: no importa la escala de la métrica, solo quién gana y por cuánto relativamente.
- #### **Aprovecha la consistencia**: ganar en muchos pares con diferencias moderadas pesa más que ganar en uno solo con diferencia enorme.


````python
import numpy as np
from scipy.stats import wilcoxon
from scipy import stats

# ------------------------------------------------------------
# Datos pareados (mismos folds de validación cruzada)
# ------------------------------------------------------------
# Cada posición i representa el MISMO fold evaluado con dos
# configuraciones distintas:
#   - sin_scaler[i]: accuracy del modelo SIN estandarización
#   - con_scaler[i]: accuracy del modelo CON StandardScaler
#
# Como ambos valores provienen del mismo split de datos,
# las observaciones son PAREADAS (dependientes), y por eso
# la prueba adecuada es wilcoxon (o ttest_rel).
#

sin_scaler = np.array([0.812, 0.798, 0.823, 0.805, 0.817,
                       0.799, 0.810, 0.821, 0.808, 0.815])

con_scaler = np.array([0.815, 0.801, 0.826, 0.808, 0.820,
                       0.802, 0.813, 0.824, 0.811, 0.980])

# ------------------------------------------------------------
# Calcular diferencias pareadas
# ------------------------------------------------------------
# d_i = con_scaler_i - sin_scaler_i mide la mejora del scaler
# en el fold i. Si el scaler ayuda, d_i > 0.
#
# Se redondea a 6 decimales para evitar problemas de precisión
# de punto flotante que podrían asignar rangos distintos a
# diferencias que deberían ser iguales (afecta a los empates
# en Wilcoxon y a la simetría en Shapiro).

# No hay regla pero si se tienen 4 decimales, se sugieren 6 decimales para no introducir empates artificiales
d = np.round(con_scaler - sin_scaler, decimals=6) 

# ------------------------------------------------------------
# 1) Supuesto de normalidad de las diferencias
# ------------------------------------------------------------
# Shapiro-Wilk contrasta H0: las diferencias provienen de una
# distribución normal.
#   - p > 0.05 → no se rechaza normalidad → ttest_rel sería válida
#   - p < 0.05 → se rechaza normalidad → usar wilcoxon
#
# Nota: el supuesto de ttest_rel es sobre las DIFERENCIAS, no
# sobre cada muestra por separado.
shapiro_stat, shapiro_p = stats.shapiro(d)
print(f"Shapiro: W={shapiro_stat:.4f}, p={shapiro_p:.4f}")

# ------------------------------------------------------------
# 2) Prueba de Wilcoxon (unilateral)
# ------------------------------------------------------------
# Hipótesis:
#   H0: la accuracy con scaler NO tiende a ser mayor que sin scaler
#   H1: la accuracy con scaler tiende a ser mayor que sin scaler
#
# alternative='greater' → prueba unilateral derecha.
#
# Wilcoxon trabaja con RANGOS y SIGNOS de las diferencias,
# no con sus magnitudes. Por eso es robusta a outliers:
# el outlier del fold 10 no infla el estadístico como lo haría
# ttest_rel, simplemente recibe el rango máximo (10) y cuenta
# como una diferencia positiva más.
#
# Estadístico reportado con alternative='greater': W+ = suma de
# rangos de las diferencias positivas. Máximo posible con n=10:
# W+_max = 10*11/2 = 55.
stat, p = wilcoxon(d, alternative='greater')
print(f"Estadístico: {stat}, p-valor: {p:.6f}")
````

# Interpretación de la prueba de Wilcoxon

## 1. Contexto

- #### **Diseño pareado**: los mismos 10 folds de validación cruzada, dos configuraciones (`con_scaler` vs. `sin_scaler`).
- #### **Métrica**: accuracy por fold (continua en [0, 1]).
- #### **Prueba**: `wilcoxon` con `alternative='greater'` → contrasta si la accuracy **tiende a ser mayor** con `StandardScaler`.
- #### **Uso de Wilcoxon**: alternativa no paramétrica a `ttest_rel` (se usa si las diferencias no cumplen normalidad).

## 2. Hipótesis

$$H_0: \text{la accuracy con scaler NO tiende a ser mayor que sin scaler} \quad (\mu_d \leq 0)$$

$$H_1: \text{la accuracy con scaler tiende a ser mayor que sin scaler} \quad (\mu_d > 0)$$

donde $d_i = \text{con\_scaler}_i - \text{sin\_scaler}_i$.

## 3. Resultados
* #### 3. Wilcoxon: $W^+ = 55$, $p = 0.000977$
* #### Con $\alpha = 0.05$:
    * #### $p = 0.000977 < 0.05 \quad \Rightarrow \quad \text{se rechaza } H_0$
