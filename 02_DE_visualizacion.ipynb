# 1. Histograma

### ¿Qué es?

#### Un **histograma** representa cómo se distribuyen los valores de una variable numérica. Los datos se agrupan en intervalos llamados **bins** (o intervalos), y la altura de cada barra  representa la **frecuencia** (o proporción) de observaciones en ese intervalo.

### ¿Qué buscamos?

#### - Concentración de los datos.
#### - Dispersión.
#### - Asimetría.
#### - Posibles valores extremos.
#### - Uno o varios grupos de concentración.

### ¿Cuándo usarlo?

#### - Para estudiar la **distribución** de una variable.
#### - Para detectar **bimodalidad** (dos grupos).
#### - Para verificar **normalidad** (visualmente).

## Ejemplo: Tiempo de programación

### Supongamos que medimos el tiempo necesario para completar una tarea de programación en 16 desarrolladores:

###  $$T = \{12, 15, 18, 20, 21, 22, 23, 25, 26, 27, 28, 29, 30, 32, 34, 40\}$$


````python
# ============================================================
# DATOS
# ============================================================
# Lista con los tiempos de programación (en minutos)
tiempos = [12, 15, 18, 20, 21, 22, 23, 25, 26, 27, 28, 29, 30, 32, 34, 40]

# Calcular la media de los tiempos
# np.mean() suma todos los valores y divide entre el número de observaciones
media = np.mean(tiempos)

# ============================================================
# CREAR EL HISTOGRAMA
# ============================================================

# plt.figure() crea una nueva figura (lienzo) para el gráfico
# figsize=(10, 6) define el tamaño en pulgadas: 10 de ancho × 6 de alto
plt.figure(figsize=(10, 6))

# plt.hist() dibuja el histograma
# - tiempos: los datos a graficar
# - bins=7: divide el rango de datos en 7 intervalos (barras)
# - edgecolor='black': color del borde de cada barra (para distinguirlas)
# - color='blue': color de relleno de las barras
plt.hist(tiempos, bins=7, edgecolor='black', color='blue')

# plt.title() agrega un título al gráfico
plt.title('Distribución de tiempos de programación')

# plt.xlabel() etiqueta el eje X (horizontal)
plt.xlabel('Tiempo (minutos)')

# plt.ylabel() etiqueta el eje Y (vertical)
plt.ylabel('Frecuencia')

# plt.axvline() dibuja una línea vertical en el gráfico
# - media: posición en el eje X donde se dibuja la línea
# - color='red': color de la línea
# - linestyle='--': estilo de línea (discontinua)
# - label=f'Media = {media:.1f}': etiqueta que aparecerá en la leyenda
#   f-string formatea el valor de la media con 1 decimal
plt.axvline(media, color='red', linestyle='--', label=f'Media = {media:.1f}')

# plt.legend() muestra la leyenda con las etiquetas definidas
plt.legend()

# plt.show() renderiza y muestra el gráfico
plt.show()
````

# 2. Boxplot

### ¿Qué es?

#### El **boxplot** (o diagrama de caja) permite visualizar:

#### - **Mediana:** el valor central.
#### - **$Q_1$:** primer cuartil (25% de los datos por debajo).
#### - **$Q_3$:** tercer cuartil (75% de los datos por debajo).
#### - **IQR:** rango intercuartílico, $IQR = Q_3 - Q_1$.
#### - **Bigotes:** extensión de los datos según el criterio utilizado.
#### - **Puntos aislados:** posibles valores atípicos.

### Componentes de un boxplot

| **Elemento** | **Significado** |
| :--- | :--- |
| $Q_1$ | Primer cuartil |
| Mediana | Percentil 50 |
| $Q_3$ | Tercer cuartil |
| $IQR$ | $Q_3 - Q_1$ |
| Bigotes | Extensión de los datos |
| Puntos aislados | Posibles valores atípicos |

### ¿Cuándo usarlo?

#### - Para comparar **grupos**.
#### - Para detectar **valores atípicos**.
#### - Para visualizar **centro y dispersión** simultáneamente.

## Componentes de un Boxplot: Q1, Q3, IQR y Bigotes

### El **boxplot** (o diagrama de caja y bigotes) es una representación gráfica que resume **cinco estadísticos** de un conjunto de datos en una sola imagen. Es especialmente útil porque muestra **simultáneamente** el centro, la dispersión y los posibles valores atípicos.

### Estructura visual de un boxplot

![Boxplot](./figs/fig-boxplot.png)



---

### 1. $Q_1$ — Primer Cuartil

| **Aspecto** | **Descripción** |
| :--- | :--- |
| **¿Qué es?** | El valor que divide el **25% inferior** de los datos del 75% superior. |
| **En el gráfico** | Es el **borde inferior de la caja**. |
| **¿Qué significa?** | El 25% de los datos están **por debajo** de $Q_1$. |
| **Ejemplo** | Si $Q_1 = 12.25$, significa que el 25% de los desarrolladores identificaron 12.25 defectos o menos. |

---

### 2. Mediana ($\tilde{x}$) — Segundo Cuartil ($Q_2$)

| **Aspecto** | **Descripción** |
| :--- | :--- |
| **¿Qué es?** | El valor que divide los datos en **dos mitades iguales**. |
| **En el gráfico** | Es la **línea dentro de la caja**. |
| **¿Qué significa?** | El 50% de los datos están **por debajo** de la mediana. |
| **Ejemplo** | Si la mediana es 13.00, significa que el 50% de los desarrolladores identificaron 13 defectos o menos. |

---

### 3. $Q_3$ — Tercer Cuartil

| **Aspecto** | **Descripción** |
| :--- | :--- |
| **¿Qué es?** | El valor que divide el **75% inferior** de los datos del 25% superior. |
| **En el gráfico** | Es el **borde superior de la caja**. |
| **¿Qué significa?** | El 75% de los datos están **por debajo** de $Q_3$. |
| **Ejemplo** | Si $Q_3 = 14.00$, significa que el 75% de los desarrolladores identificaron 14 defectos o menos. |

---

### 4. IQR — Rango Intercuartílico

| **Aspecto** | **Descripción** |
| :--- | :--- |
| **¿Qué es?** | La diferencia entre $Q_3$ y $Q_1$: $IQR = Q_3 - Q_1$. |
| **En el gráfico** | Es la **altura de la caja**. |
| **¿Qué significa?** | Contiene el **50% central** de los datos. |
| **Ejemplo** | Si $Q_1 = 12.25$ y $Q_3 = 14.00$, entonces $IQR = 1.75$. |

</br>
</br>

> ### **¿Por qué es importante el IQR?**
> - ### Es una medida de **dispersión robusta**: no se ve afectada por valores atípicos.
> - ### Se usa para detectar valores atípicos: cualquier dato fuera de los rangos mínimo (límite inferior) y máximo (límite superior) se considera atípico.
    >>  - ### Límite inferior: $Q_1 - 1.5 \times IQR$ 
    >>  - ### Límite superior: $Q_3 + 1.5 \times IQR$ 


---

### 5. Bigotes

| **Aspecto** | **Descripción** |
| :--- | :--- |
| **¿Qué son?** | Las líneas que se extienden desde la caja hasta el **valor más lejano que no es atípico**. |
| **En el gráfico** | Son las **líneas horizontales** en los extremos. |
| **¿Qué significan?** | Muestran el **rango de los datos no atípicos**. |
| **Criterio común** | Los bigotes se extienden hasta el valor más lejano dentro de $Q_1 - 1.5 \times IQR$ y $Q_3 + 1.5 \times IQR$. |

---



### Ejemplo concreto con datos

### **Datos:** Defectos identificados por 10 desarrolladores usando el Método A:
$$[11, 12, 12, 13, 13, 13, 14, 14, 15, 15]$$

#### Paso 1: Calcular los cuartiles

- Para calcular los cuartiles manualmente: 
    - Primero se ordenan los datos de menor a mayor. 
    - Luego se usa la fórmula de posición: $(n−1)×p/100(n−1)×p/100$. 
        - Por ejemplo, para Q1​ con n=10  y p=25, la posición es 9 × 0.25= 2.25 Esto significa que Q1​ está entre el segundo y tercer dato, considerando el indice 0. 
    - Se interpola: 12 + 0.25 × (13−12) =12.25  NumPy usa este método. 
    - Nota: El método más simple (Tukey) divide los datos en dos mitades y calcula la mediana de cada una. Ambos son válidos, pero pueden dar resultados ligeramente diferentes.

</br>


| **Estadístico** | **Cálculo** | **Valor** | **Interpretación** |
| :--- | :--- | :---: | :--- |
| **$Q_1$** | Percentil 25 | 12.25 | 25% de los datos ≤ 12.25 |
| **Mediana** | Percentil 50 | 13.00 | 50% de los datos ≤ 13.00 |
| **$Q_3$** | Percentil 75 | 14.00 | 75% de los datos ≤ 14.00 |
| **IQR** | $Q_3 - Q_1$ | 1.75 | 50% central de los datos |


````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Datos
datos = [11, 12, 12, 13, 13, 13, 14, 14, 15, 15]

# Convertir a DataFrame
df = pd.DataFrame({'defectos': datos})

print("Datos:")
print(df)

# ============================================================
# CÁLCULO DE CUARTILES
# ============================================================
# Los cuartiles dividen los datos ordenados en cuatro partes iguales.
# Q1 = 25% de los datos por debajo
# Q2 (mediana) = 50% de los datos por debajo
# Q3 = 75% de los datos por debajo

# np.percentile() calcula el valor del percentil indicado.
# Recibe dos argumentos:
#   1. datos: el conjunto de datos
#   2. percentil: el valor entre 0 y 100

# Calcular Q1 (primer cuartil, percentil 25)
# Significa: el 25% de los datos están por debajo de este valor
q1 = np.percentile(datos, 25)

# Calcular la mediana (segundo cuartil, percentil 50)
# Significa: el 50% de los datos están por debajo de este valor
mediana = np.percentile(datos, 50)

# Calcular Q3 (tercer cuartil, percentil 75)
# Significa: el 75% de los datos están por debajo de este valor
q3 = np.percentile(datos, 75)

# Calcular el IQR (rango intercuartílico)
# Es la diferencia entre Q3 y Q1.
# Representa la dispersión del 50% central de los datos.
# IQR = Q3 - Q1
iqr = q3 - q1

print("PASO 1: CUARTILES")
print(f"Q1 (Percentil 25):  {q1:.2f}")
print(f"Mediana (Percentil 50): {mediana:.2f}")
print(f"Q3 (Percentil 75):  {q3:.2f}")
print(f"IQR (Q3 - Q1):      {iqr:.2f}")
````


#### Paso 2: Calcular los límites para valores atípicos

| **Límite** | **Cálculo** | **Valor** |
| :--- | :--- | :---: |
| **Límite inferior** | $Q_1 - 1.5 \times IQR$ | $12.25 - 2.625 = 9.625$ |
| **Límite superior** | $Q_3 + 1.5 \times IQR$ | $14.00 + 2.625 = 16.625$ |



````python
# ============================================================
# CÁLCULO DE LÍMITES PARA VALORES ATÍPICOS
# ============================================================
# Usamos el criterio del IQR (rango intercuartílico) para
# detectar posibles valores atípicos.
#
# Un valor se considera posible atípico si está fuera de:
#   - Límite inferior: Q1 - 1.5 * IQR
#   - Límite superior: Q3 + 1.5 * IQR
#
# El factor 1.5 es una convención estadística propuesta por Tukey.
# ============================================================

# Calcular el límite inferior
# Fórmula: Q1 - 1.5 * IQR
# Cualquier valor MENOR que este límite es un posible atípico
limite_inferior = q1 - 1.5 * iqr

# Calcular el límite superior
# Fórmula: Q3 + 1.5 * IQR
# Cualquier valor MAYOR que este límite es un posible atípico
limite_superior = q3 + 1.5 * iqr

# ============================================================
# MOSTRAR LOS RESULTADOS
# ============================================================

# Imprimir encabezado del paso 2
print("PASO 2: LÍMITES PARA VALORES ATÍPICOS")

print(f"Límite inferior: Q1 - 1.5 * IQR = {q1:.2f} - 1.5 * {iqr:.2f} = {limite_inferior:.3f}")
print(f"Límite superior: Q3 + 1.5 * IQR = {q3:.2f} + 1.5 * {iqr:.2f} = {limite_superior:.3f}")
````


#### Paso 3: Interpretar el boxplot

| **Elemento** | **Valor** | **Interpretación** |
| :--- | :---: | :--- |
| **Mínimo** | 11 | Valor más pequeño |
| **$Q_1$** | 12.25 | 25% de los datos ≤ 12.25 |
| **Mediana** | 13.00 | 50% de los datos ≤ 13.00 |
| **$Q_3$** | 14.00 | 75% de los datos ≤ 14.00 |
| **Máximo** | 15 | Valor más grande |
| **IQR** | 1.75 | Dispersión del 50% central |
| **Valores atípicos** | Ninguno | Todos los datos están dentro de los límites |


````python
# Verificar si hay atípicos
atipicos = [d for d in datos if d < limite_inferior or d > limite_superior]
print(f"\nValores atípicos: {atipicos if atipicos else 'Ninguno'}")

# Resumen de todos los estadísticos
minimo = np.min(datos)
maximo = np.max(datos)
````

````python
# ============================================================
# CONSTRUCCIÓN DEL BOXPLOT
# ============================================================

# plt.figure() crea una nueva figura (lienzo) para el gráfico.
# figsize=(8, 6) define el tamaño en pulgadas: 8 de ancho × 6 de alto.
# Equivale a 20.32 cm × 15.24 cm.
plt.figure(figsize=(8, 6))

# sns.boxplot() dibuja el boxplot con Seaborn.
# - y=datos: los datos van en el eje Y (boxplot vertical).
# - color='steelblue': color de relleno de la caja.
# - width=0.4: ancho de la caja (0 a 1). 0.4 es más angosto que el default (0.8).
sns.boxplot(y=datos, color='steelblue', width=0.4)

# plt.title() agrega un título al gráfico.
plt.title('Boxplot de defectos identificados (Método A)')

# plt.ylabel() etiqueta el eje Y (vertical).
plt.ylabel('Número de defectos')

# plt.grid() activa la cuadrícula del gráfico.
# - True: muestra la cuadrícula.
# - alpha=0.3: transparencia de las líneas (0 = invisible, 1 = opaco).
plt.grid(True, alpha=0.3)

# ============================================================
# AÑADIR ANOTACIONES AL GRÁFICO
# ============================================================
# plt.text() añade texto en una posición específica del gráfico.
# Parámetros:
#   - x: posición horizontal (unidades del eje X).
#   - y: posición vertical (unidades del eje Y).
#   - texto: cadena a mostrar.
#   - fontsize: tamaño de la fuente.
#   - color: color del texto.

# Anotación para Q1
# x=0.22 coloca el texto a la derecha del centro de la caja.
# y=q1 coloca el texto a la altura del primer cuartil.
plt.text(0.22, q1, f'Q1 = {q1:.2f}', fontsize=10, color='darkgreen')

# Anotación para la mediana
# y=mediana coloca el texto a la altura de la mediana.
plt.text(0.22, mediana, f'Mediana = {mediana:.2f}', fontsize=10, color='darkred')

# Anotación para Q3
# y=q3 coloca el texto a la altura del tercer cuartil.
plt.text(0.22, q3, f'Q3 = {q3:.2f}', fontsize=10, color='darkgreen')

# Anotación para el mínimo
# y=minimo coloca el texto a la altura del valor mínimo.
plt.text(0.22, minimo, f'Mínimo = {minimo}', fontsize=10, color='black')

# Anotación para el máximo
# y=maximo coloca el texto a la altura del valor máximo.
plt.text(0.22, maximo, f'Máximo = {maximo}', fontsize=10, color='black')

# Anotación para indicar el 50% central de los datos
# x=-0.06 coloca el texto ligeramente a la izquierda del centro.
# y=13.5 coloca el texto entre Q1 y Q3 (a la altura del centro de la caja).
plt.text(-0.06, 13.5, f'50% de datos', fontsize=10, color='red')

# plt.show() renderiza y muestra el gráfico.
plt.show()
````

# Ejemplo: Comparación de dos métodos

### Supongamos que comparamos dos métodos de desarrollo (A y B) y medimos el tiempo de desarrollo en minutos.

````python
# ============================================================
# CREAR UN DATAFRAME CON DOS MÉTODOS
# ============================================================

# pd.DataFrame() crea una tabla (DataFrame) a partir de un diccionario.
# Cada clave del diccionario se convierte en una COLUMNA.
# Cada lista de valores se convierte en los DATOS de esa columna.
df_metodos = pd.DataFrame({
    # Columna 'metodo': contiene las etiquetas de los métodos (A o B)
    # Hay 5 valores 'A' y 5 valores 'B' → 10 filas en total
    'metodo': ['A', 'A', 'A', 'A', 'A', 'B', 'B', 'B', 'B', 'B'],

    # Columna 'tiempo': contiene los tiempos (en minutos) de cada observación
    # Los primeros 5 tiempos corresponden al Método A
    # Los últimos 5 tiempos corresponden al Método B
    'tiempo': [18, 20, 19, 21, 22, 25, 28, 24, 30, 35]
})

# Muestra el DataFrame en la consola
print(df_metodos)
````

````python
# ============================================================
# BOXPLOT COMPARATIVO CON SEABORN
# ============================================================

# plt.figure() crea una nueva figura (lienzo) para el gráfico.
# figsize=(8, 6) define el tamaño en pulgadas: 8 de ancho × 6 de alto.
# Matplotlib crea el LIENZO
plt.figure(figsize=(8, 6))

# sns.boxplot() dibuja el boxplot comparativo.
# Seaborn DIBUJA sobre ese lienzo
sns.boxplot(
    # x='metodo': la variable categórica va en el eje X.
    # Aquí se muestran los grupos 'A' y 'B' uno al lado del otro.
    x='metodo',

    # y='tiempo': la variable numérica va en el eje Y.
    # Se muestra la distribución de los tiempos para cada método.
    y='tiempo',

    # data=df_metodos: el DataFrame que contiene los datos.
    # Seaborn busca las columnas 'metodo' y 'tiempo' dentro de este DataFrame.
    data=df_metodos,

    # hue='metodo': asigna un color diferente a cada grupo del eje X.
    # Sin este parámetro, Seaborn lanza un warning
    # se usa 'palette'. Con 'hue', el color depende de la variable 'metodo'.
    hue='metodo',

    # palette='Set1': define la paleta de colores.
    # 'Set1' es una paleta con colores vivos y bien diferenciados.
    palette='Set1',
)

# plt.title() agrega un título al gráfico.
plt.title('Comparación de tiempos por método')

# plt.xlabel() etiqueta el eje X.
plt.xlabel('Método')

# plt.ylabel() etiqueta el eje Y.
plt.ylabel('Tiempo (minutos)')

# plt.show() renderiza y muestra el gráfico.
plt.show()
````

````python
# ============================================================
# Cálculo de Cuartiles
# ============================================================

import numpy as np
import pandas as pd

# ============================================================
# FILTRAR LOS DATOS DEL MÉTODO A
# ============================================================

# df_metodos es el DataFrame que contiene las columnas 'metodo' y 'tiempo'
# df_metodos['metodo'] == 'A' crea una MÁSCARA BOOLEANA:
#   - True donde el método es 'A'
#   - False donde el método es 'B'
# df_metodos[df_metodos['metodo'] == 'A'] FILTRA las filas donde la máscara es True
# ['tiempo'] SELECCIONA solo la columna 'tiempo' de esas filas
# El resultado es una Serie de Pandas con los tiempos del Método A
tiempos_a = df_metodos[df_metodos['metodo'] == 'A']['tiempo']

# ============================================================
# CALCULAR LOS CUARTILES
# ============================================================

# quantile(0.25) calcula el PRIMER CUARTIL (Q1)
# Es el valor que deja al 25% de los datos por debajo
q1 = tiempos_a.quantile(0.25)

# quantile(0.50) calcula el SEGUNDO CUARTIL (Q2), que es la MEDIANA
# Es el valor que deja al 50% de los datos por debajo
q2 = tiempos_a.quantile(0.50)

# quantile(0.75) calcula el TERCER CUARTIL (Q3)
# Es el valor que deja al 75% de los datos por debajo
q3 = tiempos_a.quantile(0.75)

# ============================================================
# CALCULAR EL IQR (RANGO INTERCUARTÍLICO)
# ============================================================

# IQR = Q3 - Q1
# Representa la dispersión del 50% central de los datos
# Es una medida ROBUSTA: no se ve afectada por valores atípicos
iqr = q3 - q1

# ============================================================
# MOSTRAR LOS RESULTADOS
# ============================================================

# Imprimir los valores calculados
print(f"Q1:  {q1}")    # Primer cuartil (25%)
print(f"Q2:  {q2}")    # Segundo cuartil (mediana, 50%)
print(f"Q3:  {q3}")    # Tercer cuartil (75%)
print(f"IQR: {iqr}")   # Rango intercuartílico (Q3 - Q1)
````

# 3. Diagrama de Dispersión


### ¿Qué es un diagrama de dispersión?

### El **diagrama de dispersión** (o *scatter plot*) es una representación gráfica que permite estudiar la **relación entre dos variables numéricas**. Cada punto en el gráfico representa una **observación** $(x_i, y_i)$.

### ¿Para qué sirve?

| **Objetivo** | **¿Qué buscamos?** |
| :--- | :--- |
| **Estudiar la relación** | ¿Las variables están asociadas? |
| **Detectar correlación** | ¿La asociación es fuerte o débil? |
| **Identificar patrones** | ¿Hay agrupaciones o tendencias? |
| **Detectar atípicos** | ¿Hay puntos alejados del patrón? |

### Patrones posibles

| **Patrón** | **Descripción** | **Ejemplo en IS** |
| :--- | :--- | :--- |
| **Tendencia positiva** | Cuando $X$ aumenta, $Y$ también | Más experiencia → más productividad |
| **Tendencia negativa** | Cuando $X$ aumenta, $Y$ disminuye | Más experiencia → menos tiempo de desarrollo |
| **Ausencia de relación** | No hay patrón evidente | Color del IDE → número de defectos |
| **Agrupaciones** | Posibles subgrupos | Dos equipos con comportamientos diferentes |
| **Valores extremos** | Posibles atípicos | Un desarrollador que tarda mucho más |
| **Relaciones no lineales** | Curvas, U, etc. | Estrés vs. rendimiento (curva U) |

</br>

> **⚠️ Advertencia fundamental:** Una asociación visual **no demuestra causalidad**. Para establecer causalidad se necesita un **experimento controlado**.

### Ejemplo: Experiencia vs. Tiempo de desarrollo

Supongamos que medimos los años de experiencia y el tiempo de desarrollo de 10 programadores.

````python
# ============================================================
# EJEMPLO: DIAGRAMA DE DISPERSIÓN
# ============================================================
# Este ejemplo muestra la relación entre años de experiencia
# y tiempo de desarrollo de 10 programadores.
# ============================================================

# ---- Datos ----
# Lista con los años de experiencia de cada programador
# Cada posición corresponde a un programador diferente
experiencia = [1, 2, 2, 3, 4, 5, 6, 7, 8, 10]

# Lista con el tiempo (en minutos) que tardó cada programador
# La posición i de 'tiempo' corresponde a la posición i de 'experiencia'
# Ejemplo: el programador con 1 año de experiencia tardó 25 minutos
tiempo = [25, 24, 23, 22, 20, 19, 18, 16, 15, 12]

# ============================================================
# CREAR EL DIAGRAMA DE DISPERSIÓN
# ============================================================

# plt.figure() crea una nueva figura (lienzo) para el gráfico.
# figsize=(10, 6) define el tamaño en pulgadas: 10 de ancho × 6 de alto.
plt.figure(figsize=(10, 6))

# plt.scatter() dibuja el diagrama de dispersión.
# - experiencia: valores del eje X (variable independiente).
# - tiempo: valores del eje Y (variable de respuesta).
# - color='steelblue': color de los puntos.
# - s=80: tamaño de los puntos (en puntos cuadrados).
# - alpha=0.7: transparencia de los puntos (0 = invisible, 1 = opaco).
plt.scatter(experiencia, tiempo, color='steelblue', s=80, alpha=0.7)

# plt.title() agrega un título al gráfico.
plt.title('Relación entre experiencia y tiempo de desarrollo')

# plt.xlabel() etiqueta el eje X (horizontal).
plt.xlabel('Años de experiencia')

# plt.ylabel() etiqueta el eje Y (vertical).
plt.ylabel('Tiempo (minutos)')

# plt.grid() activa la cuadrícula del gráfico.
# - True: muestra la cuadrícula.
# - alpha=0.3: transparencia de las líneas (0 = invisible, 1 = opaco).
#   Un valor bajo hace que la cuadrícula sea sutil y no distraiga.
plt.grid(True, alpha=0.3)

# plt.show() renderiza y muestra el gráfico.
plt.show()
````
