# Pruebas de Hipótesis: Paramétricas vs. No Paramétricas

---

## ¿Qué son las pruebas de hipótesis?

### Son procedimientos estadísticos que nos permiten **tomar decisiones** sobre una población a partir de los datos de una muestra. Nos ayudan a responder preguntas como:

> ### *"¿La nueva herramienta reduce realmente el tiempo de desarrollo, o la diferencia observada es solo producto del azar?"*

## Objetivos

### 1. Entender qué es una prueba de hipótesis.
### 2. Diferenciar las pruebas paramétricas de las no paramétricas.
### 3. Identificar cuándo usar cada una según los supuestos de los datos.
### 4. Conocer las pruebas más comunes de cada familia.

# 1. La decisión fundamental: usar una prueba ¿paramétrica o no paramétrica?

### Antes de aplicar cualquier prueba, debemos hacernos una pregunta clave: **¿mis datos cumplen los supuestos?**

| <span style="font-size: 28px;">**Tipo de prueba**</span> | <span style="font-size: 28px;">**¿Qué asume?**</span> | <span style="font-size: 28px;">**¿Qué usa?**</span> | <span style="font-size: 28px;">**¿Cuándo usarla?**</span> |
| :--- | :--- | :--- | :--- |
| <span style="font-size: 28px;"> **Paramétrica**</span> | <span style="font-size: 28px;"> Normalidad y (usualmente) homocedasticidad.</span>  | <span style="font-size: 28px;">Medias y varianzas. </span>| <span style="font-size: 28px;">Cuando los datos cumplen los supuestos.</span> |
| <span style="font-size: 28px;">**No paramétrica**</span> | <span style="font-size: 28px;">No asume una distribución específica. | <span style="font-size: 28px;">Rangos u ordenaciones.</span> | <span style="font-size: 28px;">Cuando los datos no son normales, son ordinales o hay valores atípicos.</span> |

</br>

> ### **Importante:** Si los datos cumplen los supuestos, usar la prueba paramétrica (es más potente). Si no, usa la no paramétrica (es más segura).

## 2. El Mapa de las Pruebas de Hipótesis

### La siguiente tabla resume las pruebas más comunes según el **objetivo**, el **número de grupos** y el **tipo de diseño** (independiente o pareado):

| **Objetivo del análisis** | **Número de grupos** | **Prueba Paramétrica** | **Prueba No Paramétrica** |
| :--- | :---: | :--- | :--- |
| **Comparar medias** | 1 (vs. referencia) | Prueba $t$ de una muestra | Prueba de los signos / Wilcoxon |
| **Comparar medias** | 2 (independientes) | Prueba $t$ de Student (o Welch) | **U de Mann-Whitney** |
| **Comparar medias** | 2 (pareados) | Prueba $t$ pareada | **Wilcoxon** |
| **Comparar medias** | 3+ (independientes) | **ANOVA** (o ANOVA de Welch) | **Kruskal-Wallis** |
| **Comparar medias** | 3+ (pareados) | ANOVA de medidas repetidas | **Friedman** |
| **Comparar medias** | 3+ (2 factores) | ANOVA factorial (2 vías) | Alineación de rangos (ART) |
| **Analizar relación** | 2 variables continuas | Correlación de **Pearson** | Correlación de **Spearman** |
| **Analizar asociación** | 2 variables categóricas | Prueba $\chi^2$ | Prueba exacta de Fisher |

</br>
</br>

> **Importante:** 
> #### * Si los datos cumplen los supuestos (normalidad y homocedasticidad), usar paramétricas. 
> #### * Si no los cumplen, usar no paramétricas.
> #### * Si las variables son categóricas, usar $\chi^2$.

## 3. El Criterio de Decisión: los Supuestos

### Para saber qué prueba usar, primero verificamos los supuestos con las herramientas que ya conocemos:
- ### **Shapiro-Wilk:** para la normalidad.
- ### **Levene:** para la homocedasticidad.    

# Ejemplo
### Un equipo de Ingeniería de Software está comparando dos servidores web. El Servidor A (Estable) mantiene tiempos de respuesta muy consistentes alrededor de los 50 ms. El Servidor B (Inestable) tiene un rendimiento errático: a veces responde muy rápido (20 ms) y a veces superiores (hasta 90 ms).
### Se midieron los tiempos de respuesta (en milisegundos) de 10 peticiones en cada servidor durante una hora de operación normal.

````python
# ============================================================
# FLUJO DE DECISIÓN: ¿QUÉ PRUEBA ESTADÍSTICA USAR?
# ============================================================
# Objetivo: Comprobar los supuestos de normalidad y
# homocedasticidad para decidir si usar una prueba paramétrica
# (t de Student/Welch) o una no paramétrica (Mann-Whitney).
# ============================================================

# ---- Paso 0: Importar librerías ----
# scipy.stats contiene las funciones estadísticas (Shapiro, Levene, t-test).
import scipy.stats as stats
import numpy as np

# ---- Paso 1: Definir los datos ----

# ---- Servidor A (Estable) ----
grupo_A = [48, 49, 50, 51, 52, 49, 50, 51, 50, 49]

# ---- Servidor B (Inestable) ----
grupo_B = [20, 80, 35, 90, 45, 25, 75, 30, 85, 40]



# ---- Paso 2: Comprobar Normalidad (Shapiro-Wilk) ----
# La prueba de Shapiro-Wilk evalúa si los datos siguen una distribución normal.
# H0: Los datos son normales.
# Si p > 0.05, no rechazamos H0 (los datos parecen normales).
# Si p < 0.05, rechazamos H0 (los datos NO son normales).
# Usamos "_" para ignorar el estadístico W y quedarnos solo con el valor p.
t_test_A, p_norm_A = stats.shapiro(grupo_A)
t_test_B_, p_norm_B = stats.shapiro(grupo_B)

# ---- Paso 3: Comprobar Homocedasticidad (Levene) ----
# La prueba de Levene evalúa si las varianzas de ambos grupos son iguales.
# H0: Las varianzas son iguales.
# Si p > 0.05, no rechazamos H0 (varianzas iguales).
# Si p < 0.05, rechazamos H0 (varianzas diferentes).
t_test_AB, p_levene = stats.levene(grupo_A, grupo_B)

# ---- Paso 4: Mostrar los resultados de los supuestos ----
# Imprimimos los valores p para que el estudiante los vea.
# El formato {:.4f} redondea a 4 decimales.
print(f"Normalidad A: p = {p_norm_A:.4f}")
print(f"Normalidad B: p = {p_norm_B:.4f}")
print(f"Levene: p = {p_levene:.4f}")

# ---- Paso 5: Tomar la decisión ----
# La decisión se basa en combinar los resultados de Shapiro y Levene.
# Usamos una estructura if-elif-else para cubrir los tres escenarios posibles.

# Escenario 1: Ambos grupos son normales Y las varianzas son iguales.
if p_norm_A > 0.05 and p_norm_B > 0.05 and p_levene > 0.05:
    print("\n→ Usar prueba t de Student (paramétrica).")

# Escenario 2: Ambos grupos son normales PERO las varianzas son diferentes.
elif p_norm_A > 0.05 and p_norm_B > 0.05 and p_levene < 0.05:
    print("\n→ Usar prueba t de Welch (paramétrica robusta).")

# Escenario 3: Al menos un grupo NO es normal (sin importar Levene).
else:
    print("\n→ Usar prueba Mann-Whitney (no paramétrica).")
````

# Prueba $t$

## Características y cuándo usar cada tipo de prueba $t$

### La prueba $t$ tiene tres variantes principales. 
* ### La elección depende del **diseño del experimento** (cuántos grupos y si están relacionados) y
* ### De los **supuestos** (normalidad y homocedasticidad).

### 1. Prueba $t$ de una muestra (One-Sample t-test)
*   ### **¿Qué compara?** La media de **un solo grupo** contra un **valor de referencia** ($\mu_0$).
*   ### **¿Cuándo usarla?** Cuando quieres saber si tu muestra difiere de un estándar conocido (ej. un SLA (Service Level Agreement), una especificación, un benchmark).
*   ### **Supuestos:**
    *   ### Los datos son continuos.
    *   ### Los datos siguen una distribución normal (Shapiro-Wilk).
*   ### **Ejemplo en IS:** ¿El tiempo medio de compilación de nuestro proyecto difiere del estándar de la empresa (45 segundos)?
*   ### **Hipótesis:** $H_0: \mu = \mu_0$ vs. $H_1: \mu \neq \mu_0$ (o $>$ / $<$).
*   ### **Función en Python:** `stats.ttest_1samp(datos, popmean=`$\mu_0$)
*   ### **Alternativa no paramétrica:** **Prueba de los signos** (si los datos no son normales) o **Wilcoxon de una muestra**.

### 2. Prueba $t$ para dos muestras independientes
*   ### **¿Qué compara?** Las medias de **dos grupos diferentes** (sujetos distintos).
*   ### **¿Cuándo usarla?** Cuando comparas dos condiciones asignadas a grupos separados (ej. Grupo A vs. Grupo B).
*   ### **Supuestos:**
    *   ### Datos continuos.
    *   ### Normalidad en ambos grupos.
    *   ### Homocedasticidad (varianzas iguales) $\rightarrow$ Si no se cumple, usar **Welch**.
*   ### **Ejemplo en IS:** ¿El tiempo de desarrollo difiere entre el equipo que usa IA y el que no la usa?
*   ### **Hipótesis:** $H_0: \mu_A = \mu_B$ vs. $H_1: \mu_A \neq \mu_B$.
*   ### **Función en Python:** `stats.ttest_ind(grupo_A, grupo_B, equal_var=True)` (Student) o `equal_var=False` (Welch).
*   ### **Alternativa no paramétrica:** **U de Mann-Whitney**.

### 3. Prueba $t$ pareada (Paired t-test)
*   ### **¿Qué compara?** Las medias de **dos mediciones relacionadas** (los mismos sujetos antes y después, o pares emparejados).
*   ### **¿Cuándo usarla?** Cuando cada sujeto proporciona dos mediciones (ej. Antes/Después, o con/sin herramienta en los mismos participantes, mismos algoritmos diferentes preprocesamientos).
*   ### **Supuestos:**
    *   ### Datos continuos.
    *   ### Las **diferencias** entre pares siguen una distribución normal (Shapiro-Wilk sobre las diferencias, no sobre los datos originales).
*   ### **Ejemplo en IS:** ¿El tiempo de resolución de bugs cambia en los mismos desarrolladores antes y después de usar una nueva herramienta?
*   ### **Hipótesis:** $H_0: \mu_d = 0$ (la media de las diferencias es cero) vs. $H_1: \mu_d \neq 0$.
*   ### **Función en Python:** `stats.ttest_rel(antes, despues)`
*   ### **Alternativa no paramétrica:** **Wilcoxon de rangos con signo**.

---

### Resumen de Decisión

| **Diseño** | **Prueba Paramétrica** | **Prueba No Paramétrica** | **Función en Python** |
| :--- | :--- | :--- | :--- |
| 1 grupo vs. referencia | $t$ de una muestra | Signos / Wilcoxon | `ttest_1samp` |
| 2 grupos independientes (varianzas iguales) | $t$ de Student | U de Mann-Whitney `stats.mannwhitneyu`| `ttest_ind(equal_var=True)` |
| 2 grupos independientes (varianzas desiguales) | $t$ de Welch | U de Mann-Whitney `stats.mannwhitneyu`| `ttest_ind(equal_var=False)` |
| 2 grupos pareados | $t$ pareada | Wilcoxon `stats.wilcoxon` | `ttest_rel` |

</br>
</br>

> ### **Importante:** 
> ### 1. Primero, verifica la normalidad con Shapiro-Wilk. Si falla, usa la versión no paramétrica.
> ### 2. Si es normal, verifica Levene (solo para grupos independientes). Si falla, usa Welch.
> ### 3. Si es pareado, verifica la normalidad de las **diferencias**, no de los datos originales.


# Ejemplo: Prueba $t$ de Student y Welch (2 grupos)

## Contexto del experimento
### Un equipo de QA está evaluando dos técnicas de pruebas (A y B) para detectar defectos. Se midió el **tiempo de detección del primer defecto** (en minutos) en 25 módulos independientes para cada técnica.

### El objetivo es determinar si existe una diferencia estadísticamente significativa en el tiempo medio de detección entre ambas técnicas.

### El flujo de decisión
### Para comparar las medias de estos dos grupos, seguiremos este flujo:
### 1. **Comprobar supuestos:** ¿Son normales? (Shapiro-Wilk). ¿Tienen varianzas iguales? (Levene).
### 2. **Elegir la prueba:** 
   - ### Si las varianzas son iguales $\rightarrow$ Prueba $t$ de **Student**.
   - ### Si las varianzas son diferentes $\rightarrow$ Prueba $t$ de **Welch**.
### 3. **Calcular el estadístico y el valor $p$.**
### 4. **Interpretar y concluir.**

````python
import numpy as np
import scipy.stats as stats

# Datos de ejemplo
datos_A = [
    9.828739, 13.994691, 12.565957, 8.987411, 10.842799,
    15.302873, 7.146642, 11.142175, 14.531873, 10.266519,
    10.642228, 11.810582, 14.982779, 10.722196, 11.112036,
    11.131297, 16.411860, 16.373572, 14.008108, 12.772373,
    13.474737, 14.981464, 10.128332, 14.351658, 9.492239
]

datos_B = [
    8.811242, 16.535526, 4.856596, 11.299656, 7.691226,
    10.721903, 1.992946, 3.142334, 8.500614, 16.637312,
    11.131822, 12.014230, 15.441114, 7.602318, 13.418137,
    7.973167, 3.361653, 10.045501, 14.869029, 13.692945,
    11.940848, 23.961826, 14.064561, 16.893680, 23.190717
]
````

## Paso 1: Comprobar supuestos
### Antes de elegir la prueba, verificamos la normalidad y la homocedasticidad.

````python
import numpy as np
import scipy.stats as stats


# 1. Normalidad (Shapiro-Wilk)
_, p_norm_A = stats.shapiro(datos_A)
_, p_norm_B = stats.shapiro(datos_B)

# 2. Homocedasticidad (Levene)
_, p_levene = stats.levene(datos_A, datos_B)

print("=== COMPROBACIÓN DE SUPUESTOS ===")
print(f"Normalidad A: p = {p_norm_A:.4f}")
print(f"Normalidad B: p = {p_norm_B:.4f}")
print(f"Levene (homocedasticidad): p = {p_levene:.4f}")

# 3. Decisión sobre la prueba
if p_norm_A > 0.05 and p_norm_B > 0.05:
    print("\n→ Ambos grupos son normales.")
    if p_levene < 0.05:
        print("   Levene p < 0.05: varianzas diferentes → Usar Welch.")
    else:
        print("   Levene p >= 0.05: varianzas iguales → Usar Student.")
else:
    print("\n→ Al menos un grupo no es normal → Usar Mann-Whitney.")
````

## Paso 2: La fórmula del estadístico $t$

### La prueba $t$ compara la diferencia de medias con el error estándar. La fórmula general es:

$$
\large
t = \frac{\bar{x}_1 - \bar{x}_2}{SE}
$$

### La diferencia entre **Student** y **Welch** está en cómo se calcula el denominador (el error estándar):

### Prueba  $t$ de Student (asume varianzas iguales)

### 1. Varianza combinada:

$$
\large
s_p^2 = \frac{(n_1 - 1)s_1^2 + (n_2 - 1)s_2^2}{n_1 + n_2 - 2}
$$

### 2. Error estándar:

$$
\large
SE = \sqrt{s_p^2 \left( \frac{1}{n_1} + \frac{1}{n_2} \right)}
$$

### 3. Grados de libertad:

$$
\large
df = n_1 + n_2 - 2
$$

### Prueba $t$ de Welch (no asume varianzas iguales)

### 1. Error estándar:

$$
\large
SE = \sqrt{\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}}
$$

### 2. Grados de libertad (aproximados):

$$
\large
df = \frac{\left( \frac{s_1^2}{n_1} + \frac{s_2^2}{n_2} \right)^2}{\frac{\left( \frac{s_1^2}{n_1} \right)^2}{n_1 - 1} + \frac{\left( \frac{s_2^2}{n_2} \right)^2}{n_2 - 1}}
$$

> ### **Regla de decisión:** Si Levene da $p < 0.05$, usamos Welch. Si da $p \geq 0.05$, usamos Student.

## Paso 3: Calcular la prueba en Python

### En Python, ambas pruebas se ejecutan con la misma función (`stats.ttest_ind`), solo cambia el parámetro `equal_var`.

### Prueba $t$ de Una Cola: El Parámetro `alternative`

### Por defecto, la función `stats.ttest_ind()` realiza una **prueba de dos colas** (busca diferencias en cualquier dirección). Sin embargo, si se tiene una **hipótesis direccional** justificada antes del experimento, se puede usar una **prueba de una cola** para aumentar la potencia estadística.

### Las Opciones de `alternative`

| **Valor de `alternative`** | **Tipo de prueba** | **Hipótesis Alternativa ($H_1$)** | **¿Cuándo usarla?** |
| :--- | :--- | :--- | :--- |
| `'two-sided'` (por defecto) | Dos colas | $\mu_A \neq \mu_B$ | No tienes dirección específica. |
| `'less'` | Una cola (izquierda) | $\mu_A < \mu_B$ | Esperas que el grupo A sea **menor** que el B. |
| `'greater'` | Una cola (derecha) | $\mu_A > \mu_B$ | Esperas que el grupo A sea **mayor** que el B. |

</br>

> ### **Importante:** La dirección debe justificarse **antes** de ver los datos.

## Ejemplo: Prueba $t$ de Welch (2 grupos)

### Contexto del experimento
### Un equipo de QA está evaluando dos técnicas de pruebas (A y B) para detectar defectos. Se midió el **tiempo de detección del primer defecto** (en minutos) en 25 módulos independientes para cada técnica.

### **Objetivo:** Determinar si la Técnica A tarda **menos tiempo** que la Técnica B en detectar el primer defecto.

### Hipótesis (direccional, cola izquierda)
*   ### **$H_0$:** $\mu_A \geq \mu_B$ (la Técnica A no es más rápida que la B).
*   ### **$H_1$:** $\mu_A < \mu_B$ (la Técnica A es más rápida que la B).

### Flujo de decisión
### 1. Comprobar supuestos (Shapiro-Wilk y Levene).
### 2. Si las varianzas son diferentes, usar la prueba $t$ de **Welch**.


### Como esperamos que A sea menor, usamos `alternative='less'`.

````python
# ============================================================
# PRUEBA T DE WELCH CON HIPÓTESIS DIRECCIONAL (COLA IZQUIERDA)
# ============================================================
# Objetivo: Determinar si la Técnica A tarda menos tiempo que
# la Técnica B en detectar defectos.
# H0: mu_A >= mu_B  (A no es más rápida)
# H1: mu_A < mu_B   (A es más rápida)
# ============================================================

# ---- Paso 1: Ejecutar la prueba ----
# stats.ttest_ind() realiza la prueba t para dos muestras independientes.
# Parámetros:
#   - datos_A: Primer grupo (Técnica A).
#   - datos_B: Segundo grupo (Técnica B).
#   - equal_var=False: Activa la prueba de Welch (NO asume varianzas iguales).
#     Esto es crucial porque Levene indicó que las varianzas son diferentes.
#   - alternative='less': Especifica una prueba de UNA COLA (izquierda).
#     Esto significa que nuestra H1 es mu_A < mu_B.
#     Si hubiéramos usado 'two-sided', sería una prueba de dos colas.
# Devuelve:
#   - stat_welch: El estadístico t calculado.
#   - p_welch: El valor p asociado a la prueba de una cola.
stat_welch, p_welch = stats.ttest_ind(
    datos_A, 
    datos_B, 
    equal_var=False,         # Welch: NO asume varianzas iguales
    alternative='less'       # Cola izquierda: H1: mu_A < mu_B
)

# ---- Paso 2: Mostrar los resultados ----
print("=== RESULTADOS (Prueba t de Welch) ===")
print(f"Estadístico t: {stat_welch:.4f}")
print(f"Valor p (una cola): {p_welch:.4f}")

# ---- Paso 3: Tomar la decisión ----
# Definimos el nivel de significancia.
alpha = 0.05

# Comparamos el valor p con alpha.
if p_welch < alpha:
    # Si p < 0.05, rechazamos H0.
    print(f"\nDecisión: Rechazar H0 (p = {p_welch:.4f} < {alpha}).")
    print("Conclusión: La Técnica A tarda significativamente menos tiempo")
    print("que la Técnica B en detectar el primer defecto.")
else:
    # Si p >= 0.05, no rechazamos H0.
    print(f"\nDecisión: No rechazar H0 (p = {p_welch:.4f} >= {alpha}).")
    print("Conclusión: No hay evidencia suficiente para afirmar que")
    print("la Técnica A sea más rápida que la Técnica B.")
````

## Paso 4: Interpretación de los resultados

### Observamos los resultados obtenidos:
- ### **Valor p (Welch):** 0.7119

### **Conclusión:**
- ### Si el valor $p$ de la prueba elegida es **menor a 0.05**, rechazamos $H_0$. Concluimos que existe una diferencia estadísticamente significativa en el tiempo medio de detección entre ambas técnicas.
- ### Si el valor $p$ es **mayor o igual a 0.05**, no rechazamos $H_0$. No hay evidencia suficiente para afirmar que las técnicas tengan rendimientos diferentes.

# Estadístico U de Mann-Whitney

* ### Contrasta la hipótesis nula de que la distribución subyacente de la muestra `x` es **idéntica** a la distribución subyacente de la muestra `y`. 

### Fórmulas de $U_1$ y $U_2$ por suma de rangos


$$
\large
U_1 = n_1 n_2 + \frac{n_1(n_1+1)}{2} - R_1
$$

$$
\large
U_2 = n_1 n_2 + \frac{n_2(n_2+1)}{2} - R_2
$$

---

## Donde:

| Símbolo | Representa |
| :--- | :--- |
| $U_1$ | Estadístico U de Mann-Whitney asociado al **grupo 1** (grupo $X$). |
| $U_2$ | Estadístico U de Mann-Whitney asociado al **grupo 2** (grupo $Y$). |
| $n_1$ | **Tamaño de la muestra** del grupo 1 (número de observaciones de $X$). |
| $n_2$ | **Tamaño de la muestra** del grupo 2 (número de observaciones de $Y$). |
| $n_1 n_2$ | **Número total de pares** posibles entre una observación del grupo 1 y una del grupo 2. Es el **valor máximo** que puede tomar $U$. |
| $R_1$ | **Suma de rangos** del grupo 1 en la muestra combinada (ambos grupos ordenados juntos de menor a mayor). |
| $R_2$ | **Suma de rangos** del grupo 2 en la muestra combinada (ambos grupos ordenados juntos de menor a mayor). |


### El estadístico U se define como el mínimo de U1 y U2.




# Prueba U de Mann-Whitney (alternativa no paramétrica a la prueba t de muestras independientes)

`scipy.stats.mannwhitneyu(x, y, alternative='two-sided')`

* ### Contrasta la hipótesis nula de que la distribución subyacente de la muestra `x` es **idéntica** a la distribución subyacente de la muestra `y`. 
* ### Es la versión no paramétrica de la prueba t para **muestras independientes**: se usa cuando **no se cumple el supuesto de normalidad** o cuando los datos son al menos ordinales.

> ### **Importante:** 
* ### `mannwhitneyu` es para **muestras independientes**. 
* ### Para muestras **pareadas/dependientes** se usa `scipy.stats.wilcoxon`.

## Parámetros principales

- #### **`x`**: primer conjunto de mediciones (Grupo A). Representa el primer grupo independiente.

- #### **`y`**: segundo conjunto de mediciones (Grupo B). Representa el segundo grupo independiente.

- #### **`alternative`**: define la hipótesis alternativa:
  - #### **`'two-sided'`** (predeterminado): las distribuciones **no son iguales**.
  - #### **`'less'`**: la distribución de `x` es **estocásticamente menor** que la de `y`.
  - #### **`'greater'`**: la distribución de `x` es **estocásticamente mayor** que la de `y`.

## Valor de retorno

#### Devuelve:
- #### **`statistic`**: el estadístico U de Mann-Whitney correspondiente a `x`.
- #### **`pvalue`**: el valor p según la alternativa elegida.

## Ejemplo de uso (comparación de dos algoritmos independientes)

```python

# Accuracy de algoritmo_A (Transformer) sobre 10 datasets de NLP
algoritmo_A = np.array([
    0.812, 0.798, 0.823, 0.805, 0.817,
```

````python
import numpy as np
from scipy import stats

# ============================================================
# Contexto: comparación de dos algoritmos sobre
# conjuntos de DATASETS DISTINTOS (muestras independientes)
# ============================================================
#
# Escenario:
# Un equipo de ML quiere saber si un nuevo modelo basado en
# Transformers (algoritmo_A) supera en accuracy a un modelo
# clásico basado en Gradient Boosting (algoritmo_B).
#
# Como los dos modelos están diseñados para DOMINIOS DISTINTOS,
# no se pueden evaluar sobre los mismos datasets:
#
#   - algoritmo_A (Transformer) se evaluó sobre 10 datasets
#     de PROCESAMIENTO DE LENGUAJE NATURAL (clasificación de texto).
#
#   - algoritmo_B (Gradient Boosting) se evaluó sobre 10 datasets
#     de DATOS TABULARES (clasificación estructurada).
#
# Cada valor es la accuracy media de validación cruzada (5-fold)
# reportada por cada modelo sobre cada dataset de su dominio.
#
# Cada algoritmo se midió sobre datasets DIFERENTES, las
# observaciones son INDEPENDIENTES: la accuracy del dataset i
# de A no tiene relación con la del dataset i de B.
#
# Hipótesis:
#   H0: Acc_A <= Acc_B  
#   H1: Acc_A > Acc_B 
# ============================================================

# Accuracy de algoritmo_A (Transformer) sobre 10 datasets de NLP
algoritmo_A = np.array([
    0.812, 0.798, 0.823, 0.805, 0.817,
    0.799, 0.810, 0.821, 0.808, 0.815
])

# Accuracy de algoritmo_B (Gradient Boosting) sobre 10 datasets tabulares
algoritmo_B = np.array([
    0.790, 0.775, 0.802, 0.785, 0.795,
    0.780, 0.788, 0.800, 0.792, 0.798
])

# Prueba unilateral: ¿algoritmo_A > algoritmo_B?
# (muestras independientes → Mann-Whitney U)
stat, p = stats.mannwhitneyu(algoritmo_A, algoritmo_B, alternative='greater')

print(f"U = {stat}, p-valor = {p:.6f}")
````

## Interpretación del resultado:

### 1. Contexto del experimento

- **`algoritmo_A`** y **`algoritmo_B`** son dos conjuntos de 10 mediciones de accuracy cada uno, obtenidos sobre **datasets distintos** → muestras **independientes**.

$$H_0: Acc_A <= Acc_B$$
$$H_1: Acc_A > Acc_B$$


$U = 95.5$ está al **95.5 % del máximo** → en el 95.5 % de los pares, un valor de A supera a uno de B.

### 2.  $p = 0.000333$

Es la probabilidad de observar un $U \geq 95.5$ **si $H_0$ fuera cierta**. Un valor de $0.000333$ (≈ 1 entre 3000) indica que es **muy improbable** que esa separación se deba al azar.


### 3. Decisión estadística

Con $\alpha = 0.05$:

$$
p = 0.000333 < 0.05 \quad \Rightarrow \quad \text{se rechaza } H_0
$$

### 4. **Conclusión:** hay evidencia estadísticamente significativa de que la accuracy del **Transformer (A)** es **estadísticamente mayor** que la del **Gradient Boosting (B)** en los conjuntos evaluados.
